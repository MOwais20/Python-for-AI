```python
def top_items(n, *items):
       top = sorted(items, key = lambda items:items[1], reverse=True)
       return [item for item, amount in top[:n]]
    
```

```python
print(top_items(2, ("pen", 50), ("book", 300), ("bag", 120)))
print(top_items(5, ("pen", 50), ("book", 300)))
print(top_items(3))
```

```python
class Product:
    def __init__(self,name,price,stock=0):
        self.name = name
        self.price = price
        self.stock = stock
    def restock(self,qty):
            self.stock += qty
    def sell(self, qty):
        if qty <= self.stock:
            self.stock -= qty
            return True
        else:
            return False
    def value(self):
        return self.price * self.stock
```

```python
pen = Product("pen", 50, 10)
print(pen.sell(4), pen.stock)
print(pen.sell(15), pen.stock)
pen.restock(4)
print(pen.value())
```
