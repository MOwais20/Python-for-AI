# 2. Getting one item out — indexing

Every item has a **position number**, called an index.

**Python counts positions from 0.** Not 1. This is the thing to burn in today.

````python
names = ["Ali", "Fatima", "Hassan", "Ayesha"]

#         0        1          2         3      <- the index

print(names[0])
print(names[1])
print(names[3])
````

````python
# BREAKS ON PURPOSE - there are 4 items, so the last index is 3
names = ["Ali", "Fatima", "Hassan", "Ayesha"]
print(names[4])
````

**IndexError** — you asked for a position that doesn't exist.

Four items means indexes 0, 1, 2, 3. The last index is always `len(list) - 1`.

### Counting from the back

Negative indexes count backwards, and `-1` is the last item. This is genuinely useful.

````python
names = ["Ali", "Fatima", "Hassan", "Ayesha"]

print(names[1::])     # last
print(names[-2])     # second to last
print(names[-4])     # first
````

> `names[-1]` is how you get the last item **without knowing how long the list is**. You will use it constantly.

````python
# Lists can hold anything - even mixed types
mixed = ["Ali", 25, 99.5, True, None]

for item in mixed:
    print(item, "->", type(item))
````

# 3. Slicing — taking a piece out

`list[start:stop]` gives you a **new** list from `start` up to **but not including** `stop`.

Yes — the same exclusive-stop rule as `range()` yesterday. Python is consistent about this.

````python
nums = [10, 20, 30, 40, 50, 60]
#        0   1   2   3   4   5

print(nums[1:4])      # positions 1, 2, 3
print(nums[:3])       # from the start
print(nums[3:])       # to the end
print(nums[:])        # the whole thing - a copy
print(nums[-2:])      # last two
````

````python
# Slicing works on text too - text is a sequence of characters
word = "Pakistan"

print(word[0])
print(word[0:4])
print(word[-3:])
print(word[::-1])     # step of -1 - reverses it
````

`[::-1]` is the classic one-line reverse. It reads as "whole thing, stepping backwards."

# 5. The list methods you will actually use

A **method** is a function that belongs to a thing. You call it with a dot: `mylist.append(...)`.

````python
fruits = ["apple", "banana"]

fruits.append("mango")            # add to the END
print(fruits)

fruits.insert(1, "orange")        # add at a position
print(fruits)

fruits.remove("banana")           # remove by VALUE
print(fruits)

last = fruits.pop()               # remove the last AND give it back
print(last, "was removed ->", fruits)
````

````python
nums = [45, 12, 88, 3, 67]

nums.sort()                       # sorts IN PLACE
print(nums)

nums.sort(reverse=True)
print(nums)

nums.reverse()
print(nums)
````

### The trap in `.sort()`

Notice it changed `nums` and returned nothing. Watch:

````python
nums = [45, 12, 88, 3, 67]

result = nums.sort()

print("nums is  :", nums)      # sorted
print("result is:", result)    # None!
````

> **`.sort()` returns `None`.** It sorts the list and hands you back nothing. `sorted_list = my_list.sort()` is a bug you will write at least once.

If you want a **new** sorted list and to keep the original, use `sorted()`:

````python
nums = [45, 12, 88, 3, 67]

new_list = sorted(nums)

print("original:", nums)
print("new     :", new_list)
````

````python
# Other things worth knowing
nums = [45, 12, 88, 3, 67, 12]

print(len(nums))           # how many
print(max(nums))           # biggest
print(min(nums))           # smallest
print(sum(nums))           # total
print(nums.count(12))      # how many times 12 appears
print(nums.index(88))      # at what position is 88
print(12 in nums)          # is 12 in there at all?
print(99 in nums)
````

# 6. The copy trap

Remember Day 1's sticker idea? It comes back to bite here, and this catches everybody.

````python
a = [1, 2, 3]
b = a              # b is a SECOND STICKER on the SAME list

b.append(4)

print("a is", a)   # a changed too!
print("b is", b)
````

`b = a` never copies an object. With a list, both names refer to the same list, so editing it through either name is visible through both.

Numbers cannot be edited in place. Assigning a different number to `b` changes what `b` refers to, without changing `a`.

To make a separate outer list:

````python
a = [1, 2, 3]
b = a.copy()       # or: b = a[:]  or  b = list(a)

b.append(4)

print("a is", a)   # unchanged
print("b is", b)
````

> **This will bite you in real projects.** Whenever you pass a list around and something mysteriously changes, this is the first thing to check.

# 7. Tuples — fixed sequences

A tuple stores items in order. It is **immutable**: you cannot replace, add, or remove its items.

We usually write tuples with round brackets: `(10, 20)`. A list stored inside a tuple can still be edited; for now, practise with numbers and words.

````python
point = (10, 20)

print(point)
print(type(point))
print(point[0])       # indexing works the same
print(len(point))
````

````python
# BREAKS ON PURPOSE - tuples are immutable
point = (10, 20)
point[0] = 99
````

### Why use a tuple?

- Coordinates, RGB colours, and fixed records are common examples.
- A tuple tells the reader that its sequence of items is not meant to change.
- A tuple can be a dictionary key if all its items are hashable, such as numbers or strings. A tuple containing a list cannot be a key. We will discuss dictionaries later.

> **Rule of thumb:** use a list when items need to change; use a tuple for a fixed sequence. Choose based on the task rather than assuming tuples are always faster.

````python
# The real reason you will meet tuples: functions that return several things
def get_name_and_age():
    return "Ali", 25          # this is a tuple

result = get_name_and_age()
print(result, type(result))
````

# 11. List push, pop, and all list methods

**Push** means adding an item to a stack. Python lists use `append()` for this action; they do not have a `push()` method. `pop()` removes and returns the last item. Think of a stack of plates: the last plate added is the first removed.

This is called **last in, first out**. See the [Python list and stack tutorial](https://docs.python.org/3/tutorial/datastructures.html#using-lists-as-stacks).

````python
stack = []  # Start with an empty stack
print("Start:", stack)

stack.append("first plate")  # Push: add one item at the end
print("After first push:", stack)

stack.append("second plate")  # Push another item
print("After second push:", stack)

removed = stack.pop()  # Pop: remove and return the last item
print("Removed:", removed)
print("Stack now:", stack)

if len(stack) > 0:  # Check before popping; an empty list would raise IndexError
    removed = stack.pop()  # Remove the remaining plate
    print("Removed:", removed)
    print("Stack now:", stack)
````

### Add one item, several items, or an item at a position

Read each operation, then check how the list changed.

````python
fruits = ["apple"]  # Make a list with one item
print("Start:", fruits)

fruits.append("banana")  # Add one item to the end
print("After append:", fruits)

fruits.extend(["mango", "orange"])  # Add each item from another list
print("After extend:", fruits)

fruits.insert(1, "grape")  # Add grape before the item at index 1
print("After insert:", fruits)
````

````python
first = [1, 2]
first.append([3, 4])  # Add the whole list as one item
print("append adds one item:", first)  # [1, 2, [3, 4]]

second = [1, 2]
second.extend([3, 4])  # Add 3 and 4 as separate items
print("extend adds separate items:", second)  # [1, 2, 3, 4]
````

### Remove by value, by position, or remove everything

`remove(value)` deletes the first match. `pop(index)` removes a position and gives back its value. `clear()` empties the list.

````python
fruits = ["apple", "banana", "mango", "banana"]
print("Start:", fruits)

fruits.remove("banana")  # Remove only the first banana
print("After remove:", fruits)

removed = fruits.pop(0)  # Remove the item at index 0
print("Removed at index 0:", removed)
print("After pop:", fruits)

fruits.clear()  # Remove every remaining item
print("After clear:", fruits)  # []
````

### Count and find

`count()` returns 0 for a missing value. `index()` and `remove()` raise `ValueError` for a missing value, so check with `in` first when necessary.

````python
numbers = [10, 20, 10, 30]
print("Numbers:", numbers)
print("How many 10s:", numbers.count(10))  # Two matches
print("How many 99s:", numbers.count(99))  # No matches
print("First position of 10:", numbers.index(10))  # Index 0

if 99 in numbers:  # Only find its position if it exists
    print("Position of 99:", numbers.index(99))
else:
    print("99 is not in the list")
````

### Sort, reverse, and copy

Sorting puts values in order. Reversing turns the current order around. Copying makes a separate outer list; this example uses plain numbers.

````python
numbers = [30, 10, 20]
print("Start:", numbers)

numbers.sort()  # Change the list to ascending order
print("After sort:", numbers)

numbers.reverse()  # Reverse the current order
print("After reverse:", numbers)

copied = numbers.copy()  # Make a separate outer list
print("Copy:", copied)

copied.append(99)  # Edit only the copy
print("Changed copy:", copied)
print("Original:", numbers)
````

### All 11 list methods at a glance

# 12. Tuple methods and the functions we use

Tuples have **only two named methods**: `count()` and `index()`. Functions such as `len()` and `sum()` also work with tuples. A function is written as `len(numbers)`; a method is written as `numbers.count(10)`.

Tuples have no push, `append()`, `pop()`, or `remove()` method.

````python
empty = ()  # A tuple with no items
print("Empty tuple:", empty)

single = (5,)  # The comma makes this a one-item tuple
print("One-item tuple:", single)
print("Type:", type(single))  # tuple

number = (5)  # Without a comma, this is just the number 5
print("Number:", number)
print("Type:", type(number))  # int
````

````python
numbers = (10, 20, 10, 30)  # A tuple with repeated values
print("Tuple:", numbers)
print("Count of 10:", numbers.count(10))  # 10 occurs twice
print("Count of 99:", numbers.count(99))  # Missing values give 0
print("First position of 10:", numbers.index(10))  # First match is index 0
print("Next position of 10:", numbers.index(10, 1))  # Start searching at index 1

if 99 in numbers:  # Avoid index() on a missing value
    print("Position of 99:", numbers.index(99))
else:
    print("99 was not found")
````

`index(value, start, stop)` can search part of a list or tuple. It stops before `stop` and returns a position in the original collection. This option can wait until the basic form feels comfortable.

````python
numbers = (10, 20, 30, 40)  # Numeric items allow totals and comparisons
print("First:", numbers[0])  # Read index 0
print("Last:", numbers[-1])  # Read from the end
print("Middle two:", numbers[1:3])  # Read indexes 1 and 2
print("Contains 20:", 20 in numbers)  # Check for a value
print("Length:", len(numbers))  # Count items
print("Total:", sum(numbers))  # Add items
print("Smallest:", min(numbers))  # Find the smallest item
print("Largest:", max(numbers))  # Find the largest item

sorted_numbers = sorted(numbers, reverse=True)  # Make a sorted list
print("Sorted result:", sorted_numbers)
print("Result type:", type(sorted_numbers))  # sorted() returns a list

backwards = tuple(reversed(numbers))  # Collect reversed items into a tuple
print("Backwards tuple:", backwards)

for position, value in enumerate(numbers):  # Read each position and item
    print("Position:", position, "Value:", value)
````

````python
point = (10, 20)
x, y = point  # Unpack two items into two names
print("x:", x)
print("y:", y)

joined = point + (30,)  # Make a new tuple containing one extra item
print("Joined tuple:", joined)
print("Original tuple:", point)  # The original did not change

repeated = point * 2  # Repeat items; do not multiply their values
print("Repeated tuple:", repeated)

editable = list(point)  # Convert to a list when edits are needed
print("As a list:", editable)
editable.append(30)  # Add an item to the list
print("After append:", editable)
new_tuple = tuple(editable)  # Create a new tuple from the edited list
print("New tuple:", new_tuple)
````

### List and tuple checklist

Use `len`, `sum`, `min`, `max`, `sorted`, `reversed`, `enumerate`, `list`, `tuple`, and `type` as shown above. Use indexing to read one item, slicing to read a part, and `in` to check a value.

`min()` and `max()` need at least one item in these examples. An average also needs an empty-input check before division. Tuples keep their item positions fixed; lists support edits.

# 13. A first look at NumPy

**NumPy** means **Numerical Python**. It is an extra Python library for working with numerical data. Its main object is an **array**: think of a row or table of numbers that you can calculate with together.

Why does it exist? Numerical work often means applying the same calculation to thousands or millions of values. NumPy provides concise operations backed by optimized code. For suitable numerical tasks it can be faster and use less memory than Python lists; it is not automatically better for every small task.

In AI, arrays can represent input measurements, image pixel values, and groups of examples. NumPy supports arithmetic, statistics, filtering, reshaping, random-number generation, and linear algebra. Today we will try only a few small examples. See [What is NumPy?](https://numpy.org/doc/stable/user/whatisnumpy.html).

### A. Add 5 to every mark

First use a list and a loop. Then do the same calculation with an array. The answers match; NumPy handles the repeated arithmetic internally.

````python
marks = [60, 70, 80]
improved_list = []  # Collect the new marks

for mark in marks:  # Work on one mark at a time
    new_mark = mark + 5  # Add 5 to this mark
    print("Old mark:", mark, "New mark:", new_mark)
    improved_list.append(new_mark)  # Keep the result

print("List result:", improved_list)

marks_array = np.array(marks)  # Convert the original marks into an array
improved_array = marks_array + 5  # Add 5 to every item
print("Array result:", improved_array)
print("Original array:", marks_array)  # This expression did not edit the original
````

### B. Lists and arrays use arithmetic differently

For lists, `+` joins and `*` repeats. For arrays, these operations calculate with the items. See the [NumPy arithmetic examples](https://numpy.org/doc/stable/user/quickstart.html#basic-operations).

````python
numbers = [1, 2, 3]
print("List times 2:", numbers * 2)  # [1, 2, 3, 1, 2, 3]

array = np.array(numbers)
print("Array times 2:", array * 2)  # [2, 4, 6]
print("Array plus 10:", array + 10)  # [11, 12, 13]
print("Array divided by 2:", array / 2)  # [0.5, 1.0, 1.5]

first = np.array([1, 2, 3])
second = np.array([10, 20, 30])  # Use the same shape for this example
print("Add matching items:", first + second)  # [11, 22, 33]
print("Multiply matching items:", first * second)  # [10, 40, 90]
````

### C. Useful calculations

An array can calculate its total, average, smallest value, and largest value. The examples below are nonempty numeric arrays.

````python
marks = np.array([60, 70, 80, 90])
print("Marks:", marks)
print("Total:", marks.sum())  # Add all marks
print("Average:", marks.mean())  # Add the marks and divide by their count
print("Smallest:", marks.min())  # Lowest mark
print("Largest:", marks.max())  # Highest mark
````

### D. Read, slice, and change items

Indexing starts at 0. Basic array slices share data with their original array; use `.copy()` when you need an independent slice. This differs from taking a slice of a list. See [NumPy indexing and slices](https://numpy.org/doc/stable/user/absolute_beginners.html#array-fundamentals).

````python
numbers = np.array([10, 20, 30, 40])
print("Original:", numbers)
print("First:", numbers[0])  # Read index 0
print("Last:", numbers[-1])  # Read the final item
print("Middle:", numbers[1:3])  # Read indexes 1 and 2

numbers[0] = 99  # Replace the first item
print("After changing the first item:", numbers)

piece = numbers[1:3]  # This slice shares the original array's data
piece[0] = 200  # Change the first item of the slice
print("Changed slice:", piece)
print("Original changed too:", numbers)

separate = numbers[1:3].copy()  # Copy the slice to keep edits separate
separate[0] = 500  # Change only the copy
print("Separate copy:", separate)
print("Original stays:", numbers)
````

### E. Keep values that match a condition

Break the operation into two steps: make a True/False selection, then use it to select the matching marks.

````python
marks = np.array([45, 70, 38, 90])
print("All marks:", marks)

passed = marks >= 50  # Check each mark separately
print("At least 50:", passed)  # False, True, False, True

passing_marks = marks[passed]  # Keep items where the check was True
print("Passing marks:", passing_marks)  # [70, 90]
````

### F. Rows, columns, and shape

A two-dimensional array is like a table. `shape` reports its layout as a **tuple**, `size` counts every item, and `ndim` counts dimensions. Attributes such as `shape` have no call brackets. See [NumPy array attributes](https://numpy.org/doc/stable/user/quickstart.html#the-basics).

````python
scores = np.array([[70, 80, 90], [60, 75, 85]])  # Two students, three subjects each
print("Scores table:", scores)
print("Shape:", scores.shape)  # (2, 3): two rows and three columns
print("Shape type:", type(scores.shape))  # tuple
print("Total items:", scores.size)  # 6
print("Dimensions:", scores.ndim)  # 2
print("Item data type:", scores.dtype)  # The integer type can depend on the environment
print("First student:", scores[0])  # Read the first row
print("Second student, third subject:", scores[1, 2])  # Row 1, column 2

rows, columns = scores.shape  # Unpack the shape tuple
print("Rows:", rows)
print("Columns:", columns)
````

### G. Make arrays and change their layout

`arange()` builds a number sequence. `zeros()` and `ones()` make starting values. `reshape()` changes the layout; the total item count must stay the same. You can explore more in the [NumPy beginner guide](https://numpy.org/doc/stable/user/absolute_beginners.html#how-to-create-a-basic-array).

````python
numbers = np.arange(1, 7)  # Make 1 through 6; stop before 7
print("Number sequence:", numbers)

zeros = np.zeros(3)  # Make three zeros; decimal numbers by default
print("Zeros:", zeros)

ones = np.ones(3)  # Make three ones
print("Ones:", ones)

table = numbers.reshape(2, 3)  # Arrange six numbers into two rows of three
print("Reshaped table:", table)
print("New shape:", table.shape)  # (2, 3)
````

### H. One data type and converting back

An array has one data type describing its items. Keep these first examples numeric; mixing numbers and text can turn them into text. Decimal values can cause integers to be stored as decimals too.

````python
numbers = np.array([1, 2.5, 3])  # Use a decimal type that can hold all these values
print("Array:", numbers)  # [1.0, 2.5, 3.0]
print("Item type:", numbers.dtype)  # A floating-point type

ordinary_list = numbers.tolist()  # Convert back to a Python list
print("As a list:", ordinary_list)
print("Type:", type(ordinary_list))  # list
````
