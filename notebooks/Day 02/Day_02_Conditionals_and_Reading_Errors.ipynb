# 1. Asking a question

Yesterday you met the comparison operators. Every one of them gives back the same thing: `True` or `False`.

That is the entire foundation of today.

````python
age = 20

print(age > 18)
print(age < 18)
print(age == 20)
print(age != 20)
````

Those four lines are **questions**. Python answered each one with a `bool`.

But so far, nothing *happened* because of the answer. That is what `if` is for.

# 2. `if` — do this, but only sometimes

````python
age = 15

if age >= 18:
    print("You can vote")
elif age > 16 or age <= 16.5 and age >= 16:
    print("You can apply for the voter Id soon.") 
else:
    print("You didnt qualify for the vote")

conditon = 2 + 2 >= 4
if conditon:
    print("The code executes")

# print("This line always runs")
````

Read it out loud exactly as it is written: **"if age is greater than or equal to 18, then print."**

### Two things about the shape of that code

**1. The colon `:`** at the end of the `if` line. It means "here comes the block."

**2. The indentation.** Those four spaces are not decoration. **In Python, indentation is how you say which lines belong inside the `if`.**

In most languages you would use curly braces `{ }`. Python uses whitespace instead. Get it wrong and your program breaks or, worse, quietly does the wrong thing.

````python
age = 15

if age >= 18:
    print("You can vote")        # INSIDE the if
print("This line always runs")   # OUTSIDE the if - notice no indent
````

`age` is 15, so the first print was skipped entirely. The second one ran anyway, because it was never inside the `if`.

````python
# BREAKS ON PURPOSE - Python expects an indented block after the colon
age = 20

if age >= 18:
  print("You can vote")
````

> **IndentationError** is the error you will see most often in your first month. It always means the same thing: a line is not where you told Python it would be.

# 3. `else` — otherwise, do this instead

````python
age = 15

if age >= 18:
    print("You can vote")
else:
    print("Too young to vote")
````

Exactly one of those two blocks runs. Never both. Never neither.

Change `age` to 20 and run it again. Then try 18 — and think about why that matters.

````python
age = 18

if age > 18:          # note: > not >=
    print("You can vote")
else:
    print("Too young to vote")

# An 18-year-old CAN vote. This code is wrong.
# > and >= are one character apart and a whole year different.
````

> **This is the most common logic bug in beginner code**, and Python will never warn you about it. The program runs perfectly. It just gives the wrong answer.

# 4. `elif` — when there are more than two doors

````python
marks = 72

if marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("Fail")
````

`elif` is short for **else if**.

### The rule that catches everyone

**Python checks them from the top, and stops at the first one that is `True`.** Everything after that is skipped, even if it would also have been true.

That is why order matters enormously.

````python
# BROKEN - the same code with the conditions in the wrong order
marks = 95

if marks >= 60:
    print("Grade C")
elif marks >= 70:
    print("Grade B")
elif marks >= 80:
    print("Grade A")
else:
    print("Fail")

# 95 is an A. This prints "Grade C".
# 95 >= 60 was true, Python stopped, and never looked at the rest.
````

**No error. No warning. Just a wrong answer.**

Remember this shape — you will write it again by accident in about two weeks.

# 5. `and`, `or`, `not` — joining questions

Plain English words, not symbols.

- **`and`** — both sides must be true
- **`or`** — at least one side must be true
- **`not`** — flips it to the opposite

````python
age = 20
has_id = True

print(age >= 18 and has_id)         # True and True   -> True
print(age >= 18 and not has_id)     # True and False  -> False
print(age >= 60 or has_id)          # False or True   -> True
print(not has_id)                   # flips True      -> False
````

````python
# Write it so it reads like a sentence
marks = 75
attendance = 90

if marks >= 50 and attendance >= 75:
    print("Passed")
else:
    print("Failed")
````

### A trap worth seeing now

````python
# WRONG ON PURPOSE - no error at all, which is exactly the problem
age = 22

if age == 20 or 21:
    print("Twenty-ish")

# It PRINTS. But run the next cell before you trust it.
````

````python
age = 5

if age == 20 and 21:
    print("Twenty-ish")     # age is 5!  Why did this print?

print(bool(21))             # <- here is why
````

Python read that as: **"is age equal to 20"** `or` **"is 21 a truthy value"**.

`21` is not zero, so Python treats it as `True`. The whole condition is therefore always true.

**You have to repeat the comparison on both sides:**

````python
age = 5

if age == 20 or age == 21:
    print("Twenty-ish")
else:
    print("Not twenty-ish")
````

# 6. Truthiness — what counts as "yes"

Python will accept almost anything where it expects a `True` or `False`. It just decides for itself.

### Everything is `True` except these

````python
print(bool(0))          # zero
print(bool(0.0))
print(bool(""))         # empty text
print(bool([]))         # empty list  (you meet these tomorrow)
print(bool(None))
print("---")
print(bool(1))
print(bool(-5))         # negative numbers are still True!
print(bool("hello"))
print(bool("0"))   


name = "Owais"

if name == "" or name != None: 
    print("Name not defined")
         # the TEXT "0" is not empty, so it is True
````

> **Empty means False. Zero means False. `None` means False. Everything else means True.**

The one that bites: `"0"` in quotes is text, and non-empty text is `True`.

````python
name = int(input("Type your number (or just press Enter): "))
print(name)

print("Type", type(name))


if name:

    print("Hello", name)
else:
    print("You didn't type anything")

# 'if name:' reads as 'if name has something in it'
````

# 7. Nesting — an `if` inside an `if`

````python
age = 20
has_id = True

if age >= 18:
    print("Age is fine")
    if has_id:
        print("ID is fine - you may enter")
    else:
        print("Go home and get your ID")
else:
    print("Too young")
````

The deeper indent shows the second `if` lives *inside* the first one. It is only ever reached if the outer condition was true.

### But don't overdo it

````python
# This does exactly the same job, and is easier to read
age = 20
has_id = True

if age >= 18 and has_id:
    print("You may enter")
elif age >= 18:
    print("Go home and get your ID")
else:
    print("Too young")
````

> **Rule of thumb:** if you are nesting three levels deep, stop and ask whether `and` would do it instead. Deeply nested code is where bugs hide.

# 8. Reading an error message

This is the most useful skill you will learn this month.

**An error is not a punishment. It is the computer telling you exactly where to look.**

Most beginners see red, panic, and start changing random things. You are going to read it instead.

````python
# BREAKS ON PURPOSE - read the output carefully before scrolling down
def greet(name):
    return "Hello " + name

print(greet("Ali"))
print(greet(25))
````

### How to read that

Python prints a **traceback**. It has two parts that matter:

**1. The LAST line is the actual error.**
```
TypeError: can only concatenate str (not "int") to str
```
That is your answer. *"You tried to join text to a number."*

**2. The lines above it tell you WHERE**, and you read them **bottom-up**.

The bottom-most file-and-line is the line that actually broke. Everything above it is the chain of how you got there.

> **Read the bottom line first. Then the line above the bottom. Ignore the middle until you need it.**

### The four errors you will meet this month

````python
# BREAKS ON PURPOSE - 1 of 4
print("hello"
````

**SyntaxError** — your grammar is wrong. A missing bracket, quote or colon.
Remember from Day 1: nothing runs at all. Caught before the program starts.

````python
# BREAKS ON PURPOSE - 2 of 4
my_name = "Ali"
print(my_nmae)
````

**NameError** — you used a name Python has never seen. Almost always a typo, or you forgot to run the cell above.

````python
# BREAKS ON PURPOSE - 3 of 4
age = "25"
print(age + 10)
````

**TypeError** — you mixed types that don't go together. Yesterday's `"5" + 5`.

````python
# BREAKS ON PURPOSE - 4 of 4
age = int("twenty five")
````

**ValueError** — the *type* was right but the *value* was wrong. `int()` accepts text, but only text that looks like a number.

> **TypeError vs ValueError:** wrong *kind* of thing, versus right kind but nonsense *content*.

### Your debugging routine, for the rest of your life

1. Read the **last line**. What kind of error?
2. Read the **line number**. Go there.
3. `print()` the variables just above that line. Look at what they actually contain.
4. Only then change something.

Step 3 is the one people skip, and it is the one that finds the bug.

````python
# Practise the routine on this one
price = "150"
quantity = 3

print("price is:", price, type(price))        # <- step 3 in action
print("quantity is:", quantity, type(quantity))

total = price * quantity
print("Total:", total)

# No error at all. But the answer is wrong. Why?
````

This is the dangerous kind: **no error, wrong answer**. `"150" * 3` repeated the text.

`print()` and `type()` found it in two seconds. Guessing would not have.

# 9. Putting it together

````python
print("=== Simple Login Check ===")

correct_user = "admin"
correct_pass = "1234"

user = input("Username: ")
password = input("Password: ")

if user == correct_user and password == correct_pass:
    print("Welcome,", user)
    
elif user == correct_user:
    print("Wrong password")
else:
    print("No such user")
````

````python
print("=== Grade Calculator ===")

marks = int(input("Enter marks out of 100: "))

if marks > 100 or marks < 0:
    print("That is not a valid mark")
elif marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("Fail")
````

Notice the **first** check catches nonsense input before any grading happens. Checking for bad input first is a habit worth building today.
