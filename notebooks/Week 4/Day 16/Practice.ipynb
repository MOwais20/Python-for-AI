```python
import matplotlib.pyplot as plt
```

```python
cars = ['Ford', 'BMW', 'Volvo']
years = [2015, 2018, 2020]

plt.plot(cars, years, marker='o')
plt.xlabel("Cars")
plt.ylabel("Years")
plt.legend(["Years"])
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Create the figure and add a 3D subplot
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

# Labeling your axes
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
```
