```python
q1 = df["price"].quantile(0.25)
q3 = df["price"].quantile(0.75)
iqr = q3 - q1
high = q3 + 1.5 * iqr
low = q1 - 1.5 * iqr


print("q1:", q1, "q3:", q3, "IQR:", round(iqr, 1))
print("Below", round(low,1), "is an outlier")
print("Above", round(high, 1), "is an outlier")
outliers = df[(df["price"] < low) | (df["price"] > high)]
print("Total outliers:", len(outliers))
outliers.sort_values("price", ascending=False).head()
```

```python
print("---Max price by cut---")
print(df.groupby("cut")["price"].max().round(2))
print()
print()
print("---Max price by clarity---")
print(df.groupby("clarity")["price"].max().round(2))
```
