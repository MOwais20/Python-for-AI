### The four pillars, on one page

| Pillar | In one line | Key tools today |
|---|---|---|
| **Encapsulation** | Data + its guard together, control access | `_protected`, `__private`, `@property` |
| **Abstraction** | Simple interface, hidden complexity | `abc.ABC`, `@abstractmethod` |
| **Inheritance** | A class reuses another | `class Child(Parent):` → Class 11 |
| **Polymorphism** | Same call, many behaviours | overriding (yes), overloading (use `*args`) |

Now put today's mechanics to work.

```python
class Account:
    def __init__(self, balance):
        self.__balance = balance

    @property
    def balance(self):               # read: acc.balance  (no brackets!)
        return self.__balance

    @balance.setter
    def balance(self, value):        # write: acc.balance = 500
        if value < 0:
            print("Refused: balance can't be negative")
            return
        self.__balance = value

acc = Account(5000)
print(acc.balance)       # looks like an attribute, runs the getter
acc.balance = -100       # runs the setter's guard -> refused
acc.balance = 8000       # allowed
print(acc.balance)
```

```python
from abc import ABC, abstractmethod
```

```python
# Abstraction
class paymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass
    def refund(self, amount):
        pass
    
class CashPayment(paymentMethod):
    def pay(self, amount):
        pass
    def refund(self, amount):
        pass
    
class CardPayment(paymentMethod):
    def pay(self, amount):
        return f"charged Rs.{amount} to card"
    def refund(self, amount):
        return f"amount RS.{amount} is refund"

for method in [CashPayment(), CardPayment()]:
    print(method.pay(2000))
    print(method.refund(1000))
```

```python
# inheritance

class Animal:
    def __init__(self,name):
        self.name =name
    def describe(self):
        return f"{self.name} is an animal"
    
class Dog(Animal):
    def bark(self):
        return "Woof"
    
d = Dog("Tommy")
print(d.describe())
print(d.bark())
```

```python
# Polymorphism
# different classes same method

class Animal:
    def speak(self):
        return"..."
class Dog(Animal):
    def speak(self):
        return "Woof"
class Cat(Animal):
    def speak(self):
        return "Meow"
    
for animal in [Dog(), Cat(), Animal()]:
    print(animal.speak())    

```

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print(f"Refused: {self.owner} has only Rs {self.balance}")
            return
        self.balance -= amount


class SavingsAccount(Account):        # SavingsAccount IS AN Account
    def add_interest(self, rate):
        self.balance += self.balance * rate

s = SavingsAccount("Ali", 10000)
s.deposit(2000)          # inherited from Account, no code repeated
s.add_interest(0.10)     # its own method
print(s.owner, "has Rs", s.balance)
```

```python
class CurrentAcount(Account):
    def __init__(self, owner, balance =0, overdraft=5000):
        super().__init__(owner, balance)
        self.overdraft= overdraft
    def withdraw(self, amount):
        if amount > self.balance + self.overdraft:
            print(f"refused past the Rs.{self.overdraft} overdraft limit")
            return
        self.balance -= amount
c = CurrentAcount('zara', 1000, overdraft=3000)
c.withdraw(3500)
print(c.owner, "Balance:",c.balance)
```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound"   # parent's version


class Dog(Animal):
    def speak(self):
        base = super().speak()      # run Animal.speak(), store its return value
        return base + " — Woof!" 
    
class GuardDog(Dog):
    def speak(self):
        base = super().speak()
        return base + " BARK BARK"

print(Animal("Thing").speak)
print(Dog("Tommy").speak())
print(GuardDog("Rex").speak())
```

```python
class Shape:
    def area(self):
        return 0

class Rectangle(Shape):
    def __init__(self,width, height):
        self.width = width
        self.height = height
    def area(self):
        return self.height * self.width
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14 * (self.radius ** 2)
r1 = Rectangle(20,20)
print(r1.area())
c1 = Circle(4)
print(c1.area())
```
