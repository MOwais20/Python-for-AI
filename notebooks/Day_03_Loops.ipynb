# 1. The problem loops solve

Suppose you want to print the numbers 1 to 5.

````python
print(1)
print(2)
print(3)
print(4)
print(5)
````

Fine. Now do 1 to 1000.

# 2. `for` — do this for every item

Read it out loud: **"for each number in this range, print the number."**

Notice the shape is the same as yesterday's `if`:

- a line ending in a **colon** `:`
- an **indented block** underneath

Every block in Python looks like this. You already know the shape.

````python
for letter in "Pakistan":
    print(letter)
````

Python walked through the text one character at a time.

The word `letter` is a name **you chose**. Python creates it fresh on each pass and fills it with the next item.

````python
# The name is yours to pick - these do exactly the same thing
for x in "Ali":
    print(x)

print("---")

for whatever_i_want in "Ali":
    print(whatever_i_want)
````

> **Pick a name that says what the thing is.** `for letter in name` reads well. `for x in name` does not. Future-you will thank present-you.

# 3. `range()` — making a sequence of numbers

`range()` is not a loop. It is a **number generator** you feed to a loop.

It has three forms.

````python
# One argument: start at 0, stop BEFORE the number
for i in range(5):
    print(i)
````

**`range(5)` gives 0, 1, 2, 3, 4.** Five numbers — but it starts at 0 and it does **not** include 5.

This surprises everyone. Two things to hold on to:

- Python counts from **0**, not 1
- The stop number is **excluded**

````python
# Two arguments: start, stop
for i in range(1, 6):
    print(i)
````

````python
# Three arguments: start, stop, step
for i in range(0, 21, 5):
    print(i)
````

````python
# Step can go backwards
for i in range(10, 0, -1):
    print(i)
print("Liftoff!")
````

### Off-by-one: the classic bug

````python
# You want 1 to 10. Count what actually prints.
for i in range(1, 10):
    print(i, end=" ")
````

It stopped at 9. `range(1, 10)` excludes 10.

You want `range(1, 11)`. **This is the single most common numeric bug in programming**, and it has a name: the off-by-one error.

````python
for i in range(1, 11):
    print(i, end=" ")
````

# 4. Doing real work in a loop

A loop is only useful if something **accumulates** across the passes.

````python
total = 0                      # start empty, OUTSIDE the loop

for number in range(1, 11):
    total = total + number     # runs 10 times, each time adding

print("Sum of 1 to 10 is", total)
````

Trace it by hand for the first three passes:

| pass | number | total before | total after |
|---|---|---|---|
| 1 | 1 | 0 | 1 |
| 2 | 2 | 1 | 3 |
| 3 | 3 | 3 | 6 |

**The `total = 0` must be outside the loop.** Watch what happens if it isn't:

````python
for number in range(1, 11):
    total = 0                  # WRONG - resets on every single pass
    total = total + number

print("Sum is", total)         # prints 10, not 55
````

No error. Wrong answer. Same category of bug as yesterday's `elif` ordering.

````python
# The shorthand you met on Day 1
total = 0
for number in range(1, 11):
    total += number            # same as total = total + number
print(total)
````

# 5. `while` — keep going until something changes

`for` runs a **known** number of times. `while` runs **until a condition stops being true**.

Use `for` when you know how many. Use `while` when you don't.

````python
count = 1

while count <= 5:
    print(count)
    count += 1

print("Done")
````

Read it as: **"while count is 5 or less, print it and add one."**

A `while` loop needs three things, and missing any one of them breaks it:

1. Something set up **before** the loop (`count = 1`)
2. A **condition** that is true at the start
3. Something **inside** that moves the condition toward false (`count += 1`)

### The infinite loop

Forget number 3 and your program never stops.

> **Don't run the next cell yet.** Read it, predict what happens, and find the Stop button (the square) next to the run button first.

````python
# INFINITE ON PURPOSE - press the STOP button to kill it
count = 1

while count <= 5:
    print(count)
    # count += 1      <- the line that is missing

# count never changes, so 'count <= 5' is true forever
````

You just wrote your first infinite loop. Everyone does. Now you know what it looks like and how to stop it.

### Where `while` genuinely beats `for`

````python
password = ""

while password != "python123":
    password = input("Enter the password: ")

print("Access granted")
````

You had no idea how many attempts it would take. That is exactly when `while` is the right tool.

**This shape — keep asking until valid — is one you will use constantly.** Your Month 2 chatbot is this loop.

# 6. `break`, `continue`, `pass`

Three small words that change how a loop flows.

### `break` — get out of the loop entirely, right now

````python
for number in range(1, 11):
    if number == 5:
        break
    print(number)

print("Loop ended")
````

It printed 1 to 4. The moment `number` hit 5, `break` abandoned the whole loop.

````python
# The pattern you will actually use: a chatbot loop
while True:                          # 'True' is always true - infinite ON PURPOSE
    message = input("Say something (or 'quit'): ")
    if message == "quit":
        break
    print("You said:", message)

print("Goodbye")
````

> **Look at that cell carefully.** `while True:` plus `break` is how almost every chat interface is built — including the chatbot you will write in Week 6. Loop forever, and let the user decide when to leave.

### `continue` — skip the rest of THIS pass, go to the next

````python
for number in range(1, 11):
    if number % 2 == 0:      # remember % from Day 1
        continue             # skip the even ones
    print(number)
````

`break` leaves the building. `continue` skips to the next person in the queue.

### `pass` — do nothing at all

Python refuses to accept an empty block. `pass` is a placeholder that says "I know something goes here, I haven't written it yet."

````python
# BREAKS ON PURPOSE - Python will not accept an empty block
for number in range(1, 4):

print("done")
````

````python
for number in range(1, 4):
    pass        # placeholder - lets the code run while you think

print("done")
````

# 7. Loops inside loops

````python
for table in range(2, 5):
    print("--- Table of", table, "---")
    for i in range(1, 11):
        print(table, "x", i, "=", table * i)
````

The inner loop runs **completely** for each single pass of the outer loop.

3 outer passes x 10 inner passes = 30 lines printed.

> **Warning for later:** nested loops multiply. Two loops over 1,000 items each is a million passes. That matters more than you think once you work with real data.

# 8. Two interview classics

These come up in real junior interviews. Not because they're hard — because they show whether you can hold a small idea in your head.

### Swapping two variables

````python
a = 10
b = 20

print("Before:", a, b)

a, b = b, a        # Python does it in one line

print("After: ", a, b)
````

````python
# How every other language does it - understand this too
a = 10
b = 20

temp = a        # park a somewhere safe
a = b           # overwrite a
b = temp        # put the parked value into b

print(a, b)
````

Remember Day 1's sticker idea? `temp` is just a third sticker, so you don't lose the value when you move the first one.

### FizzBuzz

The rules: print 1 to 20, but

- multiples of **3** print `Fizz`
- multiples of **5** print `Buzz`
- multiples of **both** print `FizzBuzz`

**Try it yourself first.** Genuinely — give it five minutes before you look.

````python
# Your attempt
````

````python
# One correct answer
for i in range(1, 21):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
````

**Why is the `and` check first?**

Because of yesterday's rule: `elif` stops at the first match. 15 is divisible by 3, so if you checked `i % 3` first, 15 would print `Fizz` and never reach the both-case.

That is yesterday's lesson and today's lesson in one problem. It is exactly why interviewers still ask it.

# 9. Looping smarter: `enumerate()` and `zip()`

Two small tools that make loops read better and save you from writing worse ones by hand.

````python
fruits = ["apple", "banana", "mango"]

for index, fruit in enumerate(fruits):
    print(index, "->", fruit)
````

`enumerate()` hands you the position and the value together, as a pair you unpack straight into two names. It replaces the clunkier `for i in range(len(fruits)): print(i, fruits[i])`.

**Use case: a numbered menu.** Pass `start=1` when counting from 1 reads better than counting from 0.

````python
options = ["Coffee", "Tea", "Juice"]

for number, option in enumerate(options, start=1):
    print(number, option)
````

### `zip()` — loop over two lists at the same time

Sometimes you have two related lists — names and marks — and want to walk through both together, position by position.

````python
names = ["Ali", "Fatima", "Hassan"]
marks = [85, 92, 78]

for name, mark in zip(names, marks):
    print(name, "scored", mark)
````

`zip()` pairs items up by position, one from each list. **If the lists are different lengths, it quietly stops at the shortest one** — no error, so check your lengths match when that would matter.

````python
names = ["Ali", "Fatima", "Hassan"]
marks = [85, 92]                     # one short

for name, mark in zip(names, marks):
    print(name, "scored", mark)

print("Hassan never printed - zip() stopped at the shorter list")
````

### Combining the two

`enumerate()` and `zip()` stack: wrap a `zip()` in `enumerate()` to get a position **and** a pair on every pass.

````python
names = ["Ali", "Fatima", "Hassan"]
marks = [85, 92, 78]

for rank, (name, mark) in enumerate(zip(names, marks), start=1):
    print(rank, name, mark)
````

### Solutions

Try the exercises yourself first — these are one way to solve them.

````python
# 1. Print 1 to 20
for i in range(1, 21):
    print(i)

# 2. Even numbers 1 to 50
for i in range(2, 51, 2):
    print(i)

# 3. Sum 1 to 100
total = 0
for i in range(1, 101):
    total += i
print("Sum:", total)

# 4. Multiplication table
number = int(input("Enter a number: "))
for i in range(1, 11):
    print(number, "x", i, "=", number * i)

# 5. Countdown then liftoff
for i in range(10, 0, -1):
    print(i)
print("Liftoff!")
````

````python
# 6. Ask until greater than 100
number = int(input("Enter a number greater than 100: "))
while number <= 100:
    number = int(input("Too small, try again: "))
print("Got it:", number)

# 7. Count vowels
word = input("Enter a word: ")
vowel_count = 0
for letter in word:
    if letter.lower() in "aeiou":
        vowel_count += 1
print("Vowels:", vowel_count)

# 8. Skip multiples of 4
for i in range(1, 21):
    if i % 4 == 0:
        continue
    print(i)
````

````python
# 9. enumerate() + zip(), numbered from 1
names = ["Ali", "Fatima", "Hassan"]
marks = [88, 72, 95]
for rank, (name, mark) in enumerate(zip(names, marks), start=1):
    print(f"{rank}: {name} - {mark}")
````
