```python
from openai import OpenAI
from dotenv import load_dotenv
import os
```

```python
llm_client = OpenAI(
    base_url='https://openrouter.ai/api/v1',
    api_key=os.getenv('OPEN_ROUTER_KEY'),
    
)
MODEL = "nvidia/nemotron-3.5-lightning:free"
EMB_MODEL = "nvidia/nemotron-3-embed-1b:free"
```

```python
messages=[
        {"role": "system", "content": """ i have a tea shop and you are mty assistant and your job it classify customer into specific format which i'll provide you.
         for label key = complaint | praise | question
         for "reason" key = give me in 4 to 5 words
         
         RESPONSE FORMATE:
         {"label":"Complaint", 'reason': 'cold chai, late'}
         
         """},
        {"role": "user", "content": ""}
              ]
```

```python
def ask_assistant(user_query):
    try: 
        messages.append({
            'role': 'user',
            'content': user_query,
            })
        completion = llm_client.chat.completions.create(
            model=MODEL,
            messages=messages
        )
        reply= completion.choices[0].message.content
        messages.append({
            'role': 'assistant',
            'content': reply,
        })
        
        print(reply)
        return reply
    except Exception as e:
        print(f"Error Occured: {e}")
```

```python
ask_assistant("The chai was too sugary i don't like it")
```

### About "Today's Agentic AI"

When people talk about *agentic AI* today, they're referring to a fast-growing class of AI systems that go beyond simple Q&A or content generation. Instead of just responding to prompts, **agentic AI** can:

- **Set its own goals** or follow multi-step objectives you give it  
- **Plan and reason** across several steps to complete a task  
- **Use tools** (search, code execution, APIs, spreadsheets, etc.) to get things done  
- **Reflect on results** and adjust its approach based on feedback  
- **Act autonomously** within the boundaries you set

#### What's Notable Right Now (2024–2025)
- **Framework-driven agents**: Projects like **AutoGPT**, **MetaGPT**, and **LangChain** let LLMs loop, plan, and execute tasks autonomously.
- **Built-in tool use**: Models like **GPT-4o**, **Claude 3.5 Sonnet**, and **Gemini 1.5 Pro** can natively call functions, search the web, write and run code, and interact with apps.
- **Enterprise automation**: Companies are deploying agentic AI for customer support, research acceleration, workflow orchestration, and code generation with human-in-the-loop oversight.
- **AI assistants with memory**: Systems that retain context across sessions, learn preferences, and proactively suggest or act on goals.

#### Why It Matters
Agentic AI marks a shift from *passive* AI (waiting for input) to *active* AI (taking initiative). This brings huge productivity gains but also raises important questions around:
- **Safety & alignment** (ensuring goals stay aligned with user intent)
- **Accountability** (who's responsible when an autonomous agent makes a mistake?)
- **Privacy** (how much data does an agent need to act effectively?)
- **Over-reliance** (balancing automation with human oversight)

#### Quick Example
You could tell an agentic AI: *"Research the top 3 climate-friendly cities, summarize their policies, and email the results to my boss."* A traditional AI might stop at the summary. An agentic one would search, analyze, draft the email, and send it—all while asking for confirmation before actually sending.
