# 1. The problem dictionaries solve

Store one student's details in a list:

````python
student = ["Ali", 25, 85, "Karachi"]

print(student[2])
````

What is `student[2]`? You have to **remember** that position 2 means marks. In three weeks you won't. And if someone inserts a phone number at position 1, every number in your program is now wrong.

**A dictionary lets you look things up by name.**

````python
student = {
    "name": "Ali",
    "age": 25,
    "marks": 85,
    "city": "Karachi"
}

print(student["marks"])
print(type(student))
````

Curly braces `{ }`, and each entry is a **key** and a **value**, joined by a colon.

`student["marks"]` says what it means. `student[2]` never did.

> **Think of a dictionary as your phone contacts.** You don't look up "contact number 47." You look up "Ali."

# 2. Keys and values

````python
student = {"name": "Ali", "age": 25, "marks": 85}

print(student["name"])
print(student["age"])
````

````python
# BREAKS ON PURPOSE - asking for a key that isn't there
student = {"name": "Ali", "age": 25}
print(student["phone"])
````

**KeyError** — that key does not exist. This is the dictionary version of yesterday's IndexError.

### `.get()` — asking safely

````python
student = {"name": "Ali", "age": 25}

print(student.get("phone"))                    # None instead of crashing
print(student.get("phone", "not provided"))    # your own fallback
print(student.get("name", "not provided"))
````

> **Use `.get()` whenever you are not certain the key is there** — which is most of the time when the data came from a file, an API, or a user. Your Month 2 AI work will be full of this.

### What can be a key?

Keys must be **immutable**. Text, numbers and tuples work. Lists do not — and now yesterday's mutability lesson pays off.

````python
# BREAKS ON PURPOSE - a list cannot be a dictionary key
d = {["a", "b"]: "value"}
````

````python
# A tuple can, because it cannot change
locations = {
    (24.86, 67.01): "Karachi",
    (31.55, 74.34): "Lahore"
}
print(locations[(24.86, 67.01)])
````

# 3. Changing a dictionary

Dictionaries are **mutable**, like lists.

````python
student = {"name": "Ali", "age": 25}

student["marks"] = 85          # key doesn't exist -> it is ADDED
print(student)

student["age"] = 26            # key exists -> it is UPDATED
print(student)

del student["age"]             # remove it
print(student)

removed = student.pop("marks") # remove and hand it back
print(removed, "->", student)
````

> **Note the one square bracket does two jobs:** adding and updating look identical. That is convenient, and it also means a typo in a key name silently creates a new entry instead of erroring.

````python
student = {"name": "Ali", "marks": 85}

student["marsk"] = 90          # typo!  No error at all.

print(student)                 # now you have both 'marks' and 'marsk'
````

# 4. Looping over a dictionary

Three ways, and the third is the one you'll use most.

````python
student = {"name": "Ali", "age": 25, "marks": 85}

for key in student:            # looping gives you the KEYS
    print(key)
````

````python
for key in student.keys():
    print(key)

print("---")

for value in student.values():
    print(value)
````

````python
for key, value in student.items():      # BOTH at once - a tuple, unpacked
    print(key, "->", value)
````

`.items()` gives you a tuple per entry, and you unpack it exactly like yesterday's `for name, mark in students`.

**This is the pattern you will use most often.** Get comfortable with it today.

# 5. Nesting — real data is shaped like this

````python
students = {
    "ali": {"age": 25, "marks": 85, "city": "Karachi"},
    "fatima": {"age": 23, "marks": 92, "city": "Lahore"}
}

print(students["ali"]["marks"])
print(students["fatima"]["city"])
````

````python
for name, details in students.items():
    print(name.title(), "is", details["age"], "and scored", details["marks"])
````

> **Look at that shape and remember it.** A dictionary of dictionaries is exactly what comes back from a web API as JSON — which is Day 18, and which is what every AI API returns in Month 2. You are already reading it.

````python
# A list of dictionaries - the other shape you will meet constantly
students = [
    {"name": "Ali", "marks": 85},
    {"name": "Fatima", "marks": 92},
    {"name": "Hassan", "marks": 78}
]

for s in students:
    print(s["name"], "scored", s["marks"])

print()
for s in students:
    if s["marks"] > 80:
        print(s["name"], "did well")
````

# 6. Sets — a bag with no duplicates and no order

A set uses curly braces like a dictionary, but holds single values, not pairs.

Two rules define it:

1. **No duplicates.** Ever.
2. **No order.** There is no `set[0]`.

````python
nums = {1, 2, 3, 2, 1, 4}

print(nums)              # the duplicates simply vanished
print(len(nums))
````

````python
# BREAKS ON PURPOSE - sets have no positions
nums = {1, 2, 3}
print(nums[0])
````

### The thing sets are genuinely brilliant at

````python
nums = [1, 2, 3, 2, 4, 1, 5, 3]

unique = list(set(nums))

print("original:", nums)
print("unique  :", unique)
````

Yesterday's duplicate-finding took a loop, a counter and an `if`. `set()` does it in one word.

> **Careful:** the order is not guaranteed to be preserved. If order matters, use `sorted(set(nums))`.

````python
# Membership checks are extremely fast in a set
names = {"ali", "fatima", "hassan"}

print("ali" in names)
print("zainab" in names)
````

### Set operations

These are the ones the outline names, and they are genuinely useful.

````python
python_students = {"ali", "fatima", "hassan", "ayesha"}
ai_students     = {"fatima", "hassan", "zainab"}

print("Both courses   :", python_students & ai_students)      # intersection
print("Either course  :", python_students | ai_students)      # union
print("Python only    :", python_students - ai_students)      # difference
print("Only one course:", python_students ^ ai_students)      # symmetric difference
````

````python
# The word versions do the same thing and read better in real code
print(python_students.intersection(ai_students))
print(python_students.union(ai_students))
print(python_students.difference(ai_students))
````

### An empty set has a catch

````python
empty = {}
print(type(empty))          # dict!  Not a set.

really_empty = set()
print(type(really_empty))   # this is how you make an empty set
````

# 7. Choosing between the four

You now know all four of Python's core containers. This table is the whole week in one place.

| | Written as | Ordered | Changeable | Duplicates | Look up by |
|---|---|---|---|---|---|
| **list** | `[1, 2, 3]` | yes | yes | yes | position |
| **tuple** | `(1, 2, 3)` | yes | **no** | yes | position |
| **dict** | `{"a": 1}` | yes | yes | keys: no | **key** |
| **set** | `{1, 2, 3}` | **no** | yes | **no** | nothing — only `in` |

**How to choose, in one sentence each:**

- Things in order that will change → **list**
- A fixed record that must not change → **tuple**
- Things you look up by name → **dict**
- A unique collection, or fast "is it in there?" → **set**

# 8. The classic: a word frequency counter

This is the problem that makes dictionaries click. It also comes up in interviews constantly.

````python
text = "the cat sat on the mat the cat was happy"

words = text.split()      # split() breaks text into a list at the spaces
print(words)
````

````python
counts = {}

for word in words:
    if word in counts:
        counts[word] += 1      # seen before - add one
    else:
        counts[word] = 1       # first time - start at one

print(counts)
````

Read those five lines again slowly. That `if in / else` shape is the heart of it.

`.get()` makes it shorter, and this is the version you will see in real code:

````python
counts = {}

for word in words:
    counts[word] = counts.get(word, 0) + 1

print(counts)
````

**"Give me the current count, or 0 if you've never seen this word, then add one."** One line, same result.

````python
# Which word appeared most?
counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1

most = max(counts, key=counts.get)
print(most, "appeared", counts[most], "times")

# Sorted from most to least
for word, n in sorted(counts.items(), key=lambda pair: pair[1], reverse=True):
    print(word, n)
````

> Don't worry about `key=lambda ...` yet — that's Day 8. Just note that it exists and that sorting a dictionary by its values is a normal thing to want.

````python
# C1. Five numbers
numbers = []
for i in range(5):
    numbers.append(float(input("Enter a number: ")))
print("Largest:", max(numbers))
print("Smallest:", min(numbers))
print("Average:", sum(numbers) / len(numbers))

# C2. Count vowels
word = input("Enter a word: ")
vowels = 0
for letter in word:
    if letter.lower() in "aeiou":
        vowels += 1
print("Vowels:", vowels)

# C3. Word frequency counter
sentence = input("Enter a sentence: ")
counts = {}
for word in sentence.split():
    counts[word] = counts.get(word, 0) + 1
most_common = max(counts, key=counts.get)
print(counts)
print("Most common:", most_common)

# C4. Passing students, sorted by marks
students = [
    {"name": "Ali", "marks": 85},
    {"name": "Fatima", "marks": 42},
    {"name": "Hassan", "marks": 67},
    {"name": "Ayesha", "marks": 91}
]
passed = []
for s in students:
    if s["marks"] >= 50:
        passed.append(s)

passed.sort(key=lambda s: s["marks"])   # sort by the 'marks' key - lambda covered on Day 8
for s in passed:
    print(s["name"], s["marks"])

# C5. Fix the bugs: total was reset inside the loop (Day 3 accumulator
# bug), and the elif order meant ">= 80" could never be reached (Day 2)
marks = [85, 42, 67]
total = 0
for m in marks:
    total += m
average = total / len(marks)
if average >= 80:
    print("Excellent class")
elif average >= 50:
    print("Class passed")
````
