# 1. What does a computer actually understand?

Inside your laptop there are **billions of tiny switches**.

A switch can only be **ON** or **OFF**. That is the whole vocabulary of a computer.

Every photo, every song, every WhatsApp message you have ever sent is just a pattern of switches being on and off.

We write **ON as 1** and **OFF as 0**. This is called **binary**, and it is the only language the machine truly speaks. Another name for it is **machine code**.

````python
# The word "Hi" as the computer actually stores it
for letter in "Hi":
    print(letter, "->", format(ord(letter), '08b'))
````

So: does a programmer sit and type 1s and 0s all day?

**No. Nobody could.**

So we need a **translator**.

# 2. What is a programming language?

A programming language is a set of words and rules that:

- **look like English to you**, and
- **can be turned into 1s and 0s for the machine**

It is a **middle language**. Human on one side, switches on the other.

**Python is one of those middle languages.** There are many. Python is the one that reads most like plain English — and it is the one the entire AI world happens to use.

````python
print("Hello")
````

One line. You just gave an instruction to a machine that only understands on and off.

# 3. What is a compiler? What is an interpreter?

Imagine you wrote a book in Urdu, and you want a Japanese man to read it.

### Way 1 — a COMPILER

You hire a translator. He takes your **whole book**, sits for a week, and hands back a **complete Japanese book**. The man then reads it himself, very fast.

- You waited a week before anything happened
- If there is one mistake on page 300, the translator refuses to give you **anything at all** until you fix it

### Way 2 — an INTERPRETER

You sit next to the man with a **live translator**. You say one sentence, he translates it, the man hears it. Then the next sentence.

- Nothing is prepared in advance
- You get the first sentence **instantly**
- But a mistake on page 300 is only discovered **when you reach page 300**

### Python works mostly like Way 2.

That is why when you press run, things happen **immediately**. You never "build" anything first.

````python
# Watch what happens here. Line 1 and 2 run. Then it stops.
print("This line runs")
print("This line also runs")
print(10 / 0)          # <-- breaks HERE
print("This line never runs")
````

Notice: the first two lines **already printed** before the error appeared.

That is the live translator. It was working line by line, and it only found the problem when it got there.

# 4. But Python also has a compiler step

Here is the full journey of your code:

```
1. Your file: main.py           <- you write this, it looks like English
        |
        v
2. COMPILER                     <- reads your whole file, checks the grammar
        |
        v
3. BYTECODE                     <- a short, tidy middle language
        |
        v
4. THE PVM (Python Virtual Machine)
        |
        v
5. MACHINE CODE (1s and 0s)     <- the CPU finally acts
```

### Box 2 — the compiler

Before **anything** runs, Python quickly reads your whole file. Not to run it — just to check your **grammar**.

Forgot a bracket? It stops right here and tells you. That is called a **syntax error**.

````python
# BREAKS ON PURPOSE - a syntax error (missing closing bracket)
print("Line one")
print("Line two"
print("Line three")
````

**Look carefully at what just happened.**

`"Line one"` did **not** print. Nothing printed at all.

Compare that to the divide-by-zero cell earlier, where two lines printed before it broke.

**That is the difference:**
- **Syntax error** = caught by the compiler, *before* anything runs. Nothing happens at all.
- **Runtime error** = caught by the interpreter, *while* running. Some of your code already worked.

### Box 3 — bytecode

The compiler turns your English-looking code into a **shorter, stricter list of instructions**.

Not 1s and 0s yet — something in between. You don't write it, and you will rarely read it. It just makes the next step faster.

**Let's actually look at it.**

````python
import dis

def add_two_numbers(a, b):
    return a + b

# This shows the bytecode Python made from that tiny function
dis.dis(add_two_numbers)
````

That is bytecode. You will never write this by hand. Python made it for you, silently.

**And it saves it to disk.** Let's prove it.

````python
%%writefile my_helper.py
def greet(name):
    return "Hello " + name
````

````python
import my_helper
print(my_helper.greet("Ali"))

# Now look - Python created a folder you never asked for
import os
print(os.listdir("__pycache__"))
````

That `.pyc` file **is the bytecode**. Python made it without telling you, so that next time it doesn't have to compile again.

### Box 4 — the PVM

**PVM = Python Virtual Machine.**

It is a program that came with Python. Its entire job is to read bytecode **one instruction at a time** and tell your actual CPU what to do.

### Why does this matter?

The PVM is the reason the **exact same file** runs on Windows, on Mac, on Linux, on a phone, and on this Google server right now.

Your file never talks to the CPU directly. It talks to the **PVM** — and each system has its own PVM that knows how to talk to that machine.

**You write once. It runs everywhere.**

````python
import sys, platform

print("Python version :", sys.version.split()[0])
print("Running on     :", platform.system())
print("The PVM in use :", platform.python_implementation())
````

> **So is Python compiled or interpreted?**
>
> **Both.** It *compiles* your file into bytecode, then *interprets* that bytecode as it goes.
>
> If anyone asks you this in an interview, that is the full answer.

# 5. Variables — a name tag on a value

````python
name = "Ali"
print(name)
````

**Never read `=` as "equals". Read it as "becomes" or "points to".**

That line means: take the value `Ali`, and stick the label `name` onto it.

Now whenever you say `name`, Python looks at what that label is pointing to.

````python
a = 10
b = a        # the label b now points at the same 10
a = 20       # only the label a moved

print("a is", a)
print("b is", b)
````

`b` is still **10**.

Moving one sticker does not move the other. They were pointing at the **same value** — they were not tied to each other.

### Rules for naming a variable

- Letters, numbers and underscore only — **no spaces, no dashes**
- **Cannot start with a number:** `2age` is invalid, `age2` is fine
- **CAPITAL LETTERS MATTER.** `Name` and `name` are two completely different variables
- Some words already belong to Python: `if`, `for`, `class`, `print`, `True`

````python
name = "Ali"
Name = "Fatima"
NAME = "Hassan"

print(name)
print(Name)
print(NAME)

# Three different variables. Python sees no connection between them at all.
````

````python
# BREAKS ON PURPOSE - cannot start with a number
2age = 25
````

# 6. Python figures out the type by itself

You never have to announce what kind of thing you are storing. You just store it, and Python looks at it and decides.

`type()` asks Python: *what kind of thing is this?*

````python
x = 5
print(type(x))
````

````python
x = "hello"      # same label, completely different kind of value
print(type(x))
````

Python let you change your mind. Some languages make you promise in advance what a label will hold, **forever**. Python does not.

This is called being **dynamically typed**.

### But Python will never *guess* across types

````python
# BREAKS ON PURPOSE
print("5" + 5)
````

You have the **text** `"5"` and the **number** `5`. They look identical to you.

To Python, one is a **word** and one is a **quantity**.

Adding a word to a quantity means nothing — so Python **refuses** instead of guessing.

**This is a good thing.** A language that guesses will silently give you wrong answers, and you will never know.

````python
# Text + text = joined together
print("5" + "5")

# Number + number = added
print(5 + 5)

# Same symbol, completely different behaviour, depending on the TYPE
````

# 7. The basic types

| Type | Means | Example |
|---|---|---|
| `int` | whole number | `age = 25` |
| `float` | number with a decimal point | `price = 99.50` |
| `str` | text — always in quotes | `name = "Ali"` |
| `bool` | yes or no | `is_open = True` |
| `NoneType` | deliberately empty | `result = None` |

````python
age = 25
price = 99.50
name = "Ali"
is_student = True
grade = None

print(age, "     ->", type(age))
print(price, "  ->", type(price))
print(name, "   ->", type(name))
print(is_student, "   ->", type(is_student))
print(grade, "   ->", type(grade))
````

### Four things to burn into your memory

**1. Quotes make text.**

`"25"` is text. `25` is a number. They are not the same thing.

````python
a = 25         # a quantity - you can do maths on it
b = "25"       # a word that happens to look like a number

print(type(a))
print(type(b))

print(a * 3)   # 75
print(b * 3)   # "252525"  <- it repeated the word!
````

**2. `True` and `False` start with CAPITAL letters.**

Lowercase `true` is an error. No reason. That is just Python's spelling.

````python
# BREAKS ON PURPOSE
is_open = true
````

````python
is_open = True     # correct
print(is_open, type(is_open))
````

**3. `None` is not zero, and not empty text.**

It means: *"I made this label, but I haven't given it anything yet."*

You will see `None` constantly in AI code.

````python
a = None
b = 0
c = ""

print(a, type(a))
print(b, type(b))
print(c, type(c))

# Three completely different kinds of "nothing"
````

**4. `10` and `10.0` are different types**, even though they are the same amount.

````python
print(type(10))
print(type(10.0))

print(10 == 10.0)     # same value...
print(type(10) == type(10.0))   # ...different type
````

# 8. Operators — doing things with values

An **operator** is a symbol that does something to values. You already know most of them from school maths.

### Arithmetic operators

````python
a = 10
b = 3

print("a + b  =", a + b)      # add
print("a - b  =", a - b)      # subtract
print("a * b  =", a * b)      # multiply
print("a / b  =", a / b)      # divide
print("a // b =", a // b)     # divide and throw away the decimal
print("a % b  =", a % b)      # the REMAINDER after dividing
print("a ** b =", a ** b)     # a to the power of b
````

Three of those need explaining properly.

**`/` always gives you a decimal — even when it divides perfectly.**

````python
print(10 / 2)          # 5.0  not 5
print(type(10 / 2))    # float!

# Division ALWAYS produces a float in Python. Always. Even 10 / 2.
````

**`//` throws away the decimal part.** It does not round — it chops.

````python
print(10 / 3)      # 3.3333333333333335
print(10 // 3)     # 3   <- chopped, not rounded
print(7 // 2)      # 3
````

"How many whole times does 3 fit into 10?" Three times. That is `//`.

**`%` gives you what is left over.** This is called **modulus**.

````python
print(10 % 3)      # 1   <- 3 fits into 10 three times, 1 is left over
print(12 % 4)      # 0   <- fits perfectly, nothing left over
````

> **Why you will use `%` constantly:** if a number divided by 2 leaves nothing over, it is **even**. If it leaves 1, it is **odd**. That is how every programmer checks odd/even.

````python
number = 7
print(number % 2)      # 1 -> odd

number = 8
print(number % 2)      # 0 -> even
````

### Comparison operators

These ask a **question**. The answer is always `True` or `False` — a `bool`.

````python
a = 10
b = 3

print("a == b :", a == b)     # are they the same?
print("a != b :", a != b)     # are they different?
print("a >  b :", a >  b)     # bigger?
print("a <  b :", a <  b)     # smaller?
print("a >= b :", a >= b)     # bigger or the same?
print("a <= b :", a <= b)     # smaller or the same?
````

### The single most common beginner mistake

> **One `=` means "becomes". Two `==` means "is it the same?"**

`=` **does** something. `==` **asks** something.

````python
x = 5          # DO: stick the label x onto the value 5
print(x == 5)  # ASK: is x the same as 5?  -> True
print(x == 9)  # ASK: is x the same as 9?  -> False
````

````python
# BREAKS ON PURPOSE - you cannot ASK a question where Python expects an ACTION
5 = x
````

### Logical operators — `and`, `or`, `not`

These join questions together. Plain English words, not symbols.

- **`and`** — both must be true
- **`or`** — at least one must be true
- **`not`** — flips it to the opposite

````python
age = 20
has_id = True

print(age >= 18 and has_id)      # both true  -> True
print(age >= 18 and not has_id)  # second one flipped -> False
print(age >= 60 or has_id)       # first false, second true -> True
````

````python
# A real example you can read out loud
marks = 75
attendance = 90

passed = marks >= 50 and attendance >= 75
print("Did the student pass?", passed)
````

### Shorthand assignment

`count = count + 1` is so common that Python gives you a short way to write it.

````python
count = 10

count += 5      # same as: count = count + 5
print(count)

count -= 3      # same as: count = count - 3
print(count)

count *= 2      # same as: count = count * 2
print(count)
````

### Order of operations

Same rules you learned in school. Multiply and divide happen **before** add and subtract.

````python
print(2 + 3 * 4)       # 14, not 20 - the * happened first
print((2 + 3) * 4)     # 20 - brackets force the order
````

> **Advice:** don't memorise the priority table. **Use brackets.** They cost nothing and they make your code readable to the next person — who is usually you, three weeks later.
