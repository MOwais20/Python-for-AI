```python
def  shout(func):
    def wrapper(name):
        print("BEFORE")
        capital = func(name)
        capital = capital.replace(name, name.upper())
        print(capital)
        print("AFTER")
    return wrapper
    

@shout 
def greet(text):
    return f'Hello, {text}'
greet('hashir ali kazmi')

```

```python
def shout(func):
    def wrapper(*args):
        result = func(*args)
        return result.upper()
    return wrapper

@shout
def greet(name):
    return f"Hello, {name}!"
print(greet("Alice"))  # Output: "HELLO, ALICE!"



# self
```
