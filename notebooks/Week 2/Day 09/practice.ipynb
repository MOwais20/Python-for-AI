```python
def my_generator():
    yield 1
    yield 2
    yield 3
    
for i in my_generator():
    print(i)
```

```python
def count_up_to(n):
    i = 1
    while i<= n:
        yield i
        i+= 1
        
for number in count_up_to(5):
    print(number)
```

```python
gen = count_up_to(5)
for number in gen:
    print(number)
```

```python
def invoice_numbers():
    number =1
    while True:
        yield f"INV-{number:04d}"
        number += 1
    
invoices = invoice_numbers()
print(next(invoices))
print(next(invoices))
print(next(invoices))
print(next(invoices))
print(next(invoices))
```

```python
noOfItems = 10
i = 0
while i <= noOfItems:
    print(next(invoices))
    i+=1    
invoices.close()
    
```

```python
def batches(items, size):
    for start in range(0,len(items), size):
        yield items[start:start+size]

orders = ["A-1001","A-1002","A-1003","A-1004","A-1005","A-1006","A-1007","A-1008","A-1009","A-1010","A-1011","A-1012","A-1013","A-1014"]
for batch in batches(orders, 5):
    print("sending to couriers:", batch)
```

```python
def simple_gen():
    yield "Email"
    yield "Tobias"
    yield "Linus"


gen = simple_gen()
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
# List comprehension - creates a list
list_comp = [x * x for x in range(5)]
print(list_comp)

# Generator expression - creates a generator
gen_exp = (x * x for x in range(5))
print(gen_exp)
print(list(gen_exp))
```

```python
def greating_generator():
    while True:
        received = yield
        print("Received" ,received)
```

```python
gen = greating_generator()
next(gen)
gen.send("Hello")
gen.send("World")
gen.close()
# gen.send("This will raise an error")
```

```python
def decorator(func):
    def shout(*args, **kwargs):
        print(f"function: {func.__name__}")
        result = func(*args, **kwargs)
        return result.upper()
    return shout
```

```python
def shout(name):
    return f"Hello, {name}!"
```

```python
shout = decorator(shout)
```

```python
print(shout('muzammil'))
```

```python
# sum_digits
def sum_digits(n):
    if n == 0:
        return 0
    return n % 10 + sum_digits(n //10)
```

```python
print(sum_digits(1234))
```

```python
def isPrime(n):
    if n %2 !=0:
        return False
    else:
        for i in range(0,31,2):
            print(i)
        return True
    
```

```python
isPrime(10)
```

```python
# generator function
def evens(limit):
    while True:
        yield limit
        limit += 2
        if limit ==10:
            break
        
```

```python

evens_gen = evens(2)
```

```python
print(list(evens_gen))
```
