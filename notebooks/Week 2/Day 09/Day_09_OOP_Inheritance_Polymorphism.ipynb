# 1. Inheritance — a child class reuses a parent

Write the shared behaviour once in a **parent** (or **base**) class. A **child** class written as `class Child(Parent):` gets every one of the parent's methods for free, and can add its own.

Here `Account` holds what every account has. `SavingsAccount` *is an* `Account` — it inherits `deposit` and `withdraw` without repeating them, and adds interest on top.

````python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print(f"Refused: {self.owner} has only Rs {self.balance}")
            return
        self.balance -= amount


class SavingsAccount(Account):        # SavingsAccount IS AN Account
    def add_interest(self, rate):
        self.balance += self.balance * rate

s = SavingsAccount("Ali", 10000)
s.deposit(2000)          # inherited from Account, no code repeated
s.add_interest(0.10)     # its own method
print(s.owner, "has Rs", s.balance)
````

`SavingsAccount` never defines `deposit` or `__init__`, yet both work — Python didn't find them on the child, so it looked *up* to the parent and used those. That upward search is the heart of inheritance.

# 2. Overriding and `super()`

A child can **override** a parent method by defining one with the same name — its version wins. And `super()` lets the child call the parent's version instead of copy-pasting it.

Overdraft: money you are allowed to withdraw even after your balance reaches zero, up to a specified limit. For example, with a Rs 3,000 overdraft limit, an account with Rs 1,000 may withdraw up to Rs 4,000. **`CurrentAccount`** overrides `withdraw` to allow an overdraft up to a limit:

A `CurrentAccount` overrides `withdraw` to allow an overdraft up to a limit:

````python
class CurrentAccount(Account):
    def __init__(self, owner, balance=0, overdraft=5000):
        super().__init__(owner, balance)     # let Account set owner + balance
        self.overdraft = overdraft           # then add our own attribute

    def withdraw(self, amount):              # override: different rule
        if amount > self.balance + self.overdraft:
            print(f"Refused: past the Rs {self.overdraft} overdraft limit")
            return
        self.balance -= amount

c = CurrentAccount("Zara", 1000, overdraft=3000)
# print(c.balance)
c.withdraw(3500)  
c.withdraw(1000)          # allowed - dips into overdraft
print(c.owner, "balance:", c.balance)

print()
````

`super().__init__(owner, balance)` runs the parent's constructor, so `CurrentAccount` doesn't repeat the `owner`/`balance` lines. Forget that call and the parent's setup never runs — the attributes it creates simply don't exist:

````python
# BREAKS ON PURPOSE - __init__ overridden but super() never called
class BrokenAccount(Account):
    def __init__(self, owner):
        self.owner = owner        # forgot to set self.balance

b = BrokenAccount("Hassan")
# b.deposit(500)                    # deposit does self.balance += ... but there is no balance
````

**AttributeError: 'BrokenAccount' object has no attribute 'balance'.** When you override `__init__`, calling `super().__init__(...)` is what keeps the parent's setup intact.

## `super()` to *extend* a method, not just replace it

Sometimes you don't want to throw away what the parent does — you want to **add** to it. Call `super().method()`, capture the result, then build on top:

```
parent does its part
       ↓
   base = super().speak()   ← run parent, save what it returned
   return base + " *wags*"  ← child adds its own extra bit
```

Without `super()` you would copy-paste the parent's logic. With it, the parent owns its logic and the child only describes what's *different*.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound"   # parent's version


class Dog(Animal):
    def speak(self):
        base = super().speak()      # run Animal.speak(), store its return value
        return base + " — Woof!"   # extend it; don't replace it


class GuardDog(Dog):
    def speak(self):
        base = super().speak()      # runs Dog.speak() (which already called Animal's)
        return base + " BARK BARK!"


print(Animal("Thing").speak())
print(Dog("Tommy").speak())
print(GuardDog("Rex").speak())

# Each level stacks on the one above it:
# Animal  →  "Rex makes a sound"
# Dog     →  "Rex makes a sound — Woof!"
# GuardDog → "Rex makes a sound — Woof! BARK BARK!"
````

**What each line does:**

| Line | What happens |
|------|--------------|
| `base = super().speak()` | jumps to the parent, runs *its* `speak`, stores the string it returns |
| `return base + " — Woof!"` | child returns the parent's string with something new tacked on |

If you just wrote `return "Woof"` you'd **replace** the parent's output entirely. Capturing into `base` first means you **keep + extend** it.

The same pattern works in `__init__` (you've already seen that) and in any other method where you want the parent's work *plus* your own.

# 3. Polymorphism — same call, different behaviour

**Polymorphism** ("many shapes") means different types respond to the *same* method name in their own way. Your code calls `.speak()` and doesn't care which animal it's holding — each object supplies its own answer.

````python
class Animal:
    def __init__(self, name):
        self.name = name
    def speak(self):
        return "..."

class Dog(Animal):
    def speak(self):
        return "Woof"

class Cat(Animal):
    def speak(self):
        return "Meow"

# One loop, three types - each speaks for itself
for animal in [Dog("Tommy"), Cat("Kitty"), Animal("Thing")]:
    print(f"{animal.name} says {animal.speak()}")
````

The loop never checks *what kind* of animal it has. Add a `Cow` class tomorrow with its own `speak()` and this loop works unchanged — that's the flexibility polymorphism buys. The same idea works on the accounts: a mixed list, one loop.

````python
accounts = [SavingsAccount("Ali", 10000), CurrentAccount("Zara", 1000)]
for acc in accounts:
    acc.withdraw(2000)        # each uses ITS OWN withdraw rule
    print(acc.owner, "->", acc.balance)
````

# 4. Dunder methods: `__str__`, `__repr__`, and `__eq__`

**Dunder** ("double underscore") methods let your objects plug into Python's own syntax. You've already met one: `__init__`. Three more you'll use constantly:

| Method | Triggered by | Used for |
|--------|-------------|----------|
| `__str__` | `print(obj)`, `str(obj)` | human-readable output |
| `__repr__` | REPL, lists, `repr(obj)` | developer/debug output |
| `__eq__` | `obj1 == obj2` | custom equality check |

Without `__str__`, printing an object is useless:

````python
# WRONG ON PURPOSE - runs, but the output is meaningless
plain = Account("Ali", 5000)
print(plain)         # <__main__.Account object at 0x...>
````

````python
class Account:
    def __init__(self, owner, balance=0, number=None):
        self.owner = owner
        self.balance = balance
        self.number = number

    def __str__(self):
        return f"{self.owner}'s account: Rs {self.balance}"

    def __eq__(self, other):
        return self.number == other.number      # same account number = same account

a = Account("Ali", 5000, number="A-001")
b = Account("Ali", 9999, number="A-001")        # different balance, same number
print(a)                # __str__ kicks in
print(a == b)           # __eq__ kicks in: True, same account
````

Now `print(a)` reads like a receipt, `a == b` asks a real question, and `repr(a)` gives you something you can paste back into Python. Each dunder hooks your object into a different part of the language.

## `__repr__` — the debug twin of `__str__`

`__str__` is for the user. `__repr__` is for the developer — shown in the REPL, inside lists, and when `__str__` isn't defined.

```
print(a)          → __str__   e.g. "Ali's account: Rs 5000"
[a, b]            → __repr__  e.g. [Account('Ali',5000), Account('Sara',3000)]
repr(a)           → __repr__  same
```

Rule of thumb: `__repr__` should look like code that could recreate the object.

````python
class Account:
    def __init__(self, owner, balance=0, number=None):
        self.owner   = owner
        self.balance = balance
        self.number  = number

    def __str__(self):
        return f"{self.owner}'s account: Rs {self.balance}"   # for humans

    def __repr__(self):
        return f"Account({self.owner!r}, {self.balance}, number={self.number!r})"  # for devs

    def __eq__(self, other):
        return self.number == other.number    # == checks account number, not balance


a = Account("Ali",  5000, number="A-001")
b = Account("Ali",  9999, number="A-001")   # same number, different balance
c = Account("Sara", 3000, number="A-002")

print(a)            # __str__ : Ali's account: Rs 5000
print(repr(a))      # __repr__: Account('Ali', 5000, number='A-001')
print([a, c])       # __repr__ inside list
print(a == b)       # __eq__  : True  (same number)
print(a == c)       # __eq__  : False (different number)
````

**Key difference in one line:**

- `__str__` → *"Ali's account: Rs 5000"* — readable sentence for a user
- `__repr__` → *`Account('Ali', 5000, number='A-001')`* — looks like Python code, useful for debugging

If you only define one, define `__repr__` — Python falls back to it for both. If you define both, `print()` uses `__str__` and everything else (REPL, lists, logs) uses `__repr__`.

# 5. Real use case: one product family, many shapes

The shop sells physical goods *and* digital licences. Both are products, but they ship differently. Inheritance shares what's common; polymorphism lets the checkout treat them all the same.

````python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def delivery_note(self):
        return "Delivered"

class PhysicalProduct(Product):
    def delivery_note(self):
        return f"Ship '{self.name}' by courier"

class DigitalProduct(Product):
    def delivery_note(self):
        return f"Email licence key for '{self.name}'"

cart = [PhysicalProduct("Keyboard", 3500), DigitalProduct("Antivirus", 1200)]
for item in cart:
    print(f"Rs {item.price:>5} - {item.delivery_note()}")   # same call, right behaviour each time
````
