```python
def count_up_to(n):
    i=1
    while i<=n:
        yield i
        i+=1
gen_func= count_up_to(5)
for num in gen_func:
    print(num)
```

Q1. Create a simple decorator (Basic)
Create a decorator called show_message that prints "Function is starting" before the decorated function runs and "Function has finished" after it runs.
Apply it to a function called greet() that prints "Hello, Python!".

```python
def decorator(func):
    def wrapper(name):
        print(f"calling {func.__name__}")
        run= func(name)
        print("function finished")
        return run
    return wrapper

@decorator
def greet(name):
    return f"Namasty {name}"

greet("Neesha")
```

```python
def double(func):
    def wrapper(num):
        return func(num) *2
    return wrapper
@double
def get_num(num):
    return num
print(get_num(10))
```

Create a decorator called uppercase_result that converts the returned string into uppercase.
Your decorator must work with functions that accept arguments.

```python
def upper(func):
    def wrapper(*args, **kwargs):
        return func(*args,**kwargs).upper()
        # return result.upper()
    return wrapper

@upper
def say_hello(name):
    return f"Hello {name}"

print(say_hello("Neesha"))
```

Create a decorator called execution_timer that measures how long a function takes to execute and prints the elapsed time in seconds.

```python
import time

def execution_timer(func):
    def wrapper():
        start = time.time()
        result= func()
        print(f"starting time {time.time()-start:.3f} seconds")
        return result
    return wrapper

@execution_timer
def process_data():
    time.sleep(2)
    return "Processing complete"

print(process_data())
```

Q5. Create a decorator that allows only a specific user (Hard)
Create a decorator called admin_only that allows a function to execute only if the supplied username is "admin".

```python


def only_admin(func):
    def wrapper(name):
        if name =="admin":
            return func(name)
        else:
            return f"Access Denied"
    return wrapper

@only_admin
def delete_database(name):
    return "Deleted Database"

print(delete_database("admin"))
```
