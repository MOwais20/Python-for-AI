Write a recursive function countdown(n) that prints numbers from n down to 1. When it reaches zero, print "Done!".

```python
def countdown(n):
    if n==0:
        return "Done"
    else:
        print(n)
        return countdown(n-1)
countdown(4)
```

Create sum_to(n) that returns the sum of all integers from 1 to n.

```python
def sum_to(n):
    if n!= 0:
       return sum_to(n-1) + n
    else:
        return 0
    
print(sum_to(5))
```

Remember: (5! = 5 * 4 *3 * 2 * 1), and (0! = 1).

```python
def factorail(n):
    if n==0:
        return 1
    else:
        return n*factorail(n-1)
print(factorail(5))
```

Reverse a string

Write reverse_string(text) that reverses a string recursively without using slicing tricks such as text[::-1].

```python
def reverse_string(text):
    if text !="":
        return  reverse_string(text[1:])+text[0]
    else:
        return text
reverse_string("Dolly")
```

Count occurrences of a value

Write count_item(items, target) that counts how many times target appears in a list.

```python
def count_item(items, target):
    if target in items:
        return 1 + count_item(items[items.index(target)+1:],target)
    else:
        return 0
count_item([2,2,1,4,5],2)
```

Write fibonacci(n) that returns the value at index n in the Fibonacci sequence.
Use the convention (F(0)=0), (F(1)=1), and (F(n)=F(n-1)+F(n-2)).

```python
def fabonacci(n):
    if n<=1:
        return n
    return fabonacci(n-1) + fabonacci(n-2)
print(fabonacci(7))
```

Find the maximum number in a list

Write recursive_max(numbers) that returns the largest number without using max() or a loop.

```python
def recursive_max(numbers):
    if len(numbers)==1:
        return numbers[0]
    if numbers[0]>numbers[1]:
        return recursive_max([numbers[0]]+numbers[2:])
    else:
        return recursive_max([numbers[1]]+numbers[2:])
print(recursive_max([4,3,2,6]))
```

Flatten a nested list

Write flatten(data) that converts a list containing nested lists into one flat list. The nesting depth can vary.

```python
def flatten(data):
    if data==[]:
        return []
    if isinstance(data[0], list): #check value and type and return true or false
        return data[0] + flatten(data[1:])
    return [data[0]] + flatten(data[1:])
print(flatten([[1,2],3,[12,3,4]])) 
```
