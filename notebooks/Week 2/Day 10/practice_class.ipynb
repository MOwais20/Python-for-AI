```python
class CustomValueError(Exception):
    pass

while True:
    try:
        user_input = int(input("Enter number"))

        if (user_input <= 0 or user_input > 120):
            raise CustomValueError("Out of bound")
        else:
            print("Completed")
            break

    except ValueError:
        print("Enter only a whole Number")
        
    except CustomValueError:
        print ("Out of bound")
        
```
