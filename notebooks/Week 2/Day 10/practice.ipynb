```python
# list of marks
class Student:
    total_students = 0
    def __init__(self, name, marks):
        self.name= name
        self.marks = list(marks)
        Student.total_students +=1
    def average(self):
        total = 0
        # return sum(marks) / len(marks)
        for i in self.marks:
            total += i
        return total / len(self.marks)
    
    def grade(self):
        if self.average()>=80:
            return "A"
        elif self.average()>=70:
            return "B"
        elif self.average()>=60:
            return "C"
        else:
            return "FAIL"
```

```python
class BankAccount:
    Bank_name="HBL"
    total_accounts=0
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        BankAccount.total_accounts += 1
    def deposite(self, amount):
        self.balance +=amount
    def withdraw(self, amount):
        self.balance -= amount

    def transfer(self, BankAccount, amount):
        if self.balance > amount:
            self.withdraw(amount)
            BankAccount.deposite(amount)
            print("Amount Transfered")
        else:
            print("Insufficient Amount")
            
```

```python
A1 = BankAccount('ali', 10000)
A2 = BankAccount('sara', 2000)

print("A1 balance:",A1.balance)
print("A2 balance:",A2.balance)
A1.transfer(A2, 5000) 
print("A1 balance:",A1.balance)
print("A2 balance:",A2.balance)
```

```python
class Product:
    def __init__(self, name, price, cost):
        self.name = name
        self.price = price      # what we sell it for
        self.cost = cost        # what we paid for it

    def profit(self):
        return self.price - self.cost


class Cart:
    def __init__(self):
        self.items = []         # each cart starts empty, with its own list

    def add(self, product, qty=1):
        self.items.append((product, qty))

    def total(self):
        return sum(p.price * qty for p, qty in self.items)

    def profit(self):
        return sum(p.profit() * qty for p, qty in self.items)


mouse    = Product("Mouse", 1500, 900)
keyboard = Product("Keyboard", 3500, 2200)

cart = Cart()
cart.add(mouse, 2)
cart.add(keyboard)

print("Total charged: Rs", cart.total())
print("Shop keeps:    Rs", cart.profit())
```
