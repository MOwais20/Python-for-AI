```python
class Animal:
    def speak(self):
        return " '?' "

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

for i in [Cat(), Dog(), Animal()]:
    print(i.speak())
```

```python
class shape:
    def area(self):
        return 0

class Rectangle(shape):
    def __init__(self, width, height):
                self.width = width
                self.height = height

    def area(self):
        return self.width * self.height

class circle(shape):
    def __init__(self, radius):
     self.radius = radius
    
    def area(self):
        return 3.14 * self.radius ** 2

c = [circle(10), Rectangle(10, 20)]
for r in c:
    print(r.area())
```

```python
class Animal:
    def __init__(self, name):
            self.name = name
            
    def speak(self):
        return " '?' "

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

for i in [Cat("Kitty"), Dog("Tommy"), Animal("Thing")]:
    print(f"{i.name} says {i.speak()}")
```

```python
class Book:
    def __init__(self, author, title, isbn):
        self.author = author
        self.title =  title
        self.isbn = isbn

    def __str__(self):
        return f"{self.title} by {self.author} and its ISBN is (ISBN: {self.isbn})"

    def __eq__(self, different):
        return self.isbn == different.isbn

c  = Book("Paulo Coelho", "The Alchemist", "9780061122415")
d = Book("George Orwell", "1984", "9780061122415")
print(c)
print(c == d)
```
