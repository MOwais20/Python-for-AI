# 1. What is OOP?

Every program mixes two things: **data** (the values) and **behaviour** (the code that acts on them). The difference between coding styles is *how you organise those two*.

**The style you've used so far — procedural.** Data sits in variables; functions act on it from the outside. A student's marks live in one place; a `calculate_average(marks)` function lives somewhere else. They're only connected by you remembering to pass the right data to the right function.

**Object-oriented programming.** You bundle the data *and* the behaviour that belongs to it into one unit called an **object**. An object is a thing that **knows its own data and what it can do with it**.

Think of real things:

| Real thing | Its data (what it *has*) | Its behaviour (what it *does*) |
|---|---|---|
| A car | colour, speed, fuel | accelerate, brake, refuel |
| A bank account | owner, balance | deposit, withdraw |
| A student | name, marks | average, grade |

OOP lets you model each of these as an object in code: one thing that carries its own data *and* the actions that go with it.

### You've been using objects all along

Python is object-oriented to the core. Every string, list and dictionary you've touched is already an object — it carries data *and* methods (behaviour) you call with a dot.

```python
name = "fatima"
print(name.upper())        # the string is an object; .upper() is its behaviour

marks = [85, 92, 78]
marks.append(100)          # the list is an object; .append() is its behaviour
print(marks)
```

`name.upper()` — the *object* is `name`, the *behaviour* is `.upper()`. You never wrote `upper(name)`; the behaviour lives **on** the object. Today you stop only *using* objects and start **building your own**.

> **The four big ideas of OOP** (your roadmap): **encapsulation** (keep data and behaviour together), **inheritance** (one class reuses another), **polymorphism** (one call, many types), and **abstraction** (hide the messy details behind a simple interface). You'll meet encapsulation today; inheritance and polymorphism are Class 11.

# 2. What OOP is used for, and why it helps

**What it's used for.** Any time your program is really *about things* — accounts, users, products, orders, game characters — OOP lets you model each thing as an object. It's everywhere you're heading:

- The libraries you'll use are built from objects. In Class 8 you wrote `app = FastAPI()` — `app` is an object. Streamlit apps, database connections, and AI models (`model = SomeModel()`) are all objects you create and call methods on.
- Real systems have *many* of the same thing — a thousand users, a hundred products. One class describes them all; each user is its own object.

**Why it helps:**

1. **Organisation** — data and the code that works on it live together, not scattered across the file.
2. **Reuse** — write the shared behaviour once, build many objects from it (and, next class, let one class inherit from another).
3. **It matches how you think** — you already think in terms of "a student", "an account". OOP lets your code say the same.
4. **Safety** — the rules that protect the data (like "don't overdraw") live *with* the data, so every part of the program plays by them.
5. **It scales** — as a program grows, objects keep it in tidy, self-contained pieces instead of one giant tangle.

# 3. Classes and objects — the blueprint and the thing built from it

Two words you must not mix up:

- A **class** is the **blueprint** — the description of what every object of this kind has and can do. It's written once.
- An **object** (also called an **instance**) is an **actual thing built from that blueprint**. You can build as many as you like from one class.

**Analogy.** A class is an architect's *blueprint* for a house. The blueprint isn't a house — it's the plan. The **objects** are the actual houses built from it. Every house shares the same layout, but each has its own address and its own coat of paint. One blueprint → many houses.

Some vocabulary you'll hear all day:

| Term | Meaning |
|---|---|
| **Class** | The blueprint (`BankAccount`) |
| **Object / instance** | One thing built from it (Ali's actual account) |
| **Attribute** | A piece of data on an object (`balance`) |
| **Method** | A behaviour on an object — a function inside the class (`deposit`) |
| **Instantiation** | Building an object from a class (`BankAccount("Ali")`) |

### Why bother — the pain a class removes

You *can* describe a bank account with a `dict` and some loose functions:

```python
account = {"owner": "Ali", "balance": 5000}

def deposit(acc, amount):
    acc["balance"] += amount

deposit(account, 2000)
print(account)
```

Two things are loose here:

- The data (`account`) and the behaviour (`deposit`) are **separate**. Nothing ties `deposit` to *this* account — you must pass it in every time and hope it's the right one.
- Nothing stops a typo. Write `acc["balence"]` and Python silently creates a **new** key instead of complaining.

A **class** packages the data and its behaviour into one object, so they can never drift apart. Let's build it.

# 4. Your first class: `__init__`, attributes and `self`

Three pieces to learn:

- `__init__` is the **constructor** — Python runs it automatically the moment you build an object, to set up its starting data.
- `self` is **this particular object**. Inside the class, `self.balance` means "*this* account's own balance", kept separate from every other account.
- The values stored on `self` (`self.owner`, `self.balance`) are the object's **attributes**.

```python
class BankAccount:
    def __init__(self, owner, balance=0):     # runs when an account is created
        self.owner = owner                    # store data ON this object
        self.balance = balance

# Instantiation: build objects from the blueprint. BankAccount(...) calls __init__.
ali = BankAccount("Ali", 5000)
zara = BankAccount("Zara")                     # balance defaults to 0

print(ali.owner, ali.balance)                  # read attributes with a dot
print(zara.owner, zara.balance)
```

Trace `BankAccount("Ali", 5000)`: Python builds a fresh empty object, calls `__init__` with `self` = that new object, `owner = "Ali"`, `balance = 5000`. The two lines glue that data onto the object, and you get it back as `ali`.

`ali` and `zara` are **independent** — each has its own `self`. Change one, the other doesn't move:

```python
ali.balance = 9999
print("Ali:", ali.balance)
print("Zara:", zara.balance)   # untouched
```

# 5. Methods — behaviour that lives with the data

A **method** is a function defined *inside* the class. Its first parameter is always `self`, so it can read and change *this* object's attributes. Now `deposit` belongs to the account instead of floating around loose — this is **encapsulation**: data and behaviour, sealed together.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        self.balance -= amount

ali = BankAccount("Ali", 5000)
ali.deposit(2000)              # note: no self passed in - Python fills it with `ali`
ali.withdraw(500)
print(ali.owner, "has Rs", ali.balance)
```

`ali.deposit(2000)` is really `BankAccount.deposit(ali, 2000)` — Python passes the object before the dot in as `self`. That is the whole trick behind `self`.

But `withdraw` has a bug: it never checks whether there's enough money.

```python
# WRONG ON PURPOSE - runs fine, wrong result: the balance goes negative
ali = BankAccount("Ali", 1000)
ali.withdraw(5000)
print("Balance:", ali.balance)   # -4000, a bank that pays you to overdraw
```

The method is the right place to defend the data — the guard lives *with* the balance it protects, so every caller is covered at once. (That's benefit #4 from Section 2, made real.)

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print(f"Refused: {self.owner} tried to withdraw Rs {amount} but has Rs {self.balance}")
            return
        self.balance -= amount

ali = BankAccount("Ali", 1000)
ali.withdraw(5000)            # refused
ali.withdraw(400)             # fine
print("Balance:", ali.balance)
```

# 6. Instance attributes vs class attributes

- An **instance attribute** (`self.balance`) is set on `self`, so every object has its own copy.
- A **class attribute** is defined directly in the class body, *outside* `__init__`, and is **shared by every object** — good for things true of the whole class: the bank's name, a running count of accounts.

```python
class BankAccount:
    bank_name = "HBL"          # class attribute - shared by all accounts
    total_accounts = 0         # shared counter

    def __init__(self, owner, balance=0):
        self.owner = owner     # instance attribute - one per object
        self.balance = balance
        BankAccount.total_accounts += 1   # bump the shared counter

ali = BankAccount("Ali", 5000)
zara = BankAccount("Zara", 3000)

print(ali.bank_name, "-", zara.bank_name)        # both see the same shared value
print("Accounts opened:", BankAccount.total_accounts)
```

Change the class attribute in one place and every object sees the new value at once — that's the point of sharing it:

```python
BankAccount.bank_name = "HBL Bank Ltd"
print(ali.bank_name)     # HBL Bank Ltd
print(zara.bank_name)    # HBL Bank Ltd - same shared attribute
```

# 7. Real use case: the shop, modelled as objects

The same computer shop from Class 8 and 9 — but now a `Product` knows its own price and cost, and a `Cart` knows how to total itself. Data and behaviour travel together.

```python
class Product:
    def __init__(self, name, price, cost):
        self.name = name
        self.price = price      # what we sell it for
        self.cost = cost        # what we paid for it

    def profit(self):
        return self.price - self.cost


class Cart:
    def __init__(self):
        self.items = []         # each cart starts empty, with its own list

    def add(self, product, qty=1):
        self.items.append((product, qty))

    def total(self):
        return sum(p.price * qty for p, qty in self.items)

    def profit(self):
        return sum(p.profit() * qty for p, qty in self.items)


mouse    = Product("Mouse", 1500, 900)
keyboard = Product("Keyboard", 3500, 2200)

cart = Cart()
cart.add(mouse, 2)
cart.add(keyboard)

print("Total charged: Rs", cart.total())
print("Shop keeps:    Rs", cart.profit())
```

Notice `Cart.profit()` leans on `Product.profit()` — each object answers for its own part. That's the payoff of keeping behaviour next to data: the cart never has to know *how* a product computes profit, only that it can.

# 8. The four pillars of OOP, in detail

Everything OOP does rests on four ideas. You've already *used* the first two today; now we go deep on all four — including access modifiers, abstract classes, and the difference between overriding and overloading. Inheritance and polymorphism get a full class of their own (Class 11); here you get the complete concept plus working examples.

## Pillar 1 — Encapsulation

**Keep data and the behaviour that guards it in one object, and control who can touch that data from outside.** The `withdraw` guard was step one. Step two is **access modifiers**: marking which attributes are open and which are internal.

### Access modifiers: public, protected, private

Languages like Java or C++ have keywords — `public`, `protected`, `private`. **Python has no such keywords.** Instead it uses a **naming convention**, and Python trusts you to respect it ("we're all adults here"):

| Intent | How you write it | What it means | Enforced? |
|---|---|---|---|
| **public** | `self.balance` | Anyone may read and change it | — |
| **protected** | `self._balance` (one `_`) | "Internal: for this class and its subclasses" | Convention only |
| **private** | `self.__balance` (two `__`) | "Belongs to this class alone" | Yes — via *name mangling* |

Public is the default. A single underscore is a polite "please don't". A double underscore is the real thing:

```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner          # public: fine to read/change from anywhere
        self._bank = "HBL"          # protected: convention says "internal"
        self.__balance = balance    # private: name-mangled, hard to reach outside

    def deposit(self, amount):
        self.__balance += amount    # inside the class, __balance works normally

    def balance(self):
        return self.__balance       # controlled read access

acc = Account("Ali", 5000)
acc.deposit(2000)
print(acc.owner)         # public - works
print(acc._bank)         # protected - works, but you were warned
print(acc.balance())     # private, reached through a method - the right way
```

### When the `_` convention breaks

A single underscore is **only a polite request**. Python won't stop anyone who ignores it — so "protected" state can be clobbered from outside, silently. Here someone changes the bank on Ali's account, and Python raises no objection at all:

```python
# WRONG ON PURPOSE - nothing stops this; the "protected" bank just changes
acc._bank = "Fake Bank of Nowhere"
print(acc._bank)         # allowed - no error, no warning
```

That is the real risk of protected: it works only while everyone respects the underscore. The moment someone doesn't, it breaks **quietly** — no error tells you the rule was ignored, so the bug shows up later somewhere else.

### So why bother with `_` at all?

If Python won't enforce it, why write it? Because **`_` is a message to humans, not a lock for the computer.** It's the author telling the next person: *"this is my internal wiring — I might rename or remove it, so don't build on it; use my public methods instead."*

Think of a door marked **"Staff Only"**. It doesn't physically stop you — but now if you walk in and something breaks, that's on you. You were told. Three things it buys you:

1. **A clear line between the buttons and the machinery.** Names *without* `_` are the official controls you're meant to use. `_names` are the internals behind the panel — you can see at a glance what's safe to rely on.
2. **Freedom for the author to change things.** If everyone agrees to use `balance()` and never touch `_bank`, the author can rename or restructure `_bank` later and nobody's code breaks — because nobody was supposed to be using it.
3. **Your tools respect it.** VS Code, linters and autocomplete treat `_names` as internal — often hiding them from suggestions and warning when you reach in from outside. So the convention gets *some* teeth from your editor even though the language shrugs.

Python's philosophy is *"we're all consenting adults here"* — it marks the boundary clearly and trusts you, rather than caging you the way a hard `private` keyword would. The ladder: **no underscore** = "public, safe to use" → **`_one`** = "internal, please don't" → **`__two`** = "internal, and I'll make it awkward." Here's that last rung — a **private** `__` name, where Python actively gets in the way. Try reaching it directly:

```python
# BREAKS ON PURPOSE - a double-underscore name can't be reached from outside as-is
print(acc.__balance)
```

**AttributeError.** Behind the scenes Python **renames** `__balance` to `_Account__balance` (this is *name mangling*). It's not truly locked — `acc._Account__balance` would still reach it — but the mangling makes accidental access impossible and signals loudly: *use the methods*.

### The Pythonic way: `@property`

Getters and setters (`get_balance()`, `set_balance()`) work, but Python's cleaner tool is `@property` — it lets a method *look like* an attribute while still running your guard code:

```python
class Account:
    def __init__(self, balance):
        self.__balance = balance

    @property
    def balance(self):               # read: acc.balance  (no brackets!)
        return self.__balance

    @balance.setter
    def balance(self, value):        # write: acc.balance = 500
        if value < 0:
            print("Refused: balance can't be negative")
            return
        self.__balance = value

acc = Account(5000)
print(acc.balance)       # looks like an attribute, runs the getter
acc.balance = -100       # runs the setter's guard -> refused
acc.balance = 8000       # allowed
print(acc.balance)
```

## Pillar 2 — Abstraction

**Show a simple interface; hide the messy details behind it.** When you call `cart.total()` you don't care *how* it sums — you just ask. `len(marks)` and `"hi".upper()` do complex work behind one short name. That's abstraction you've used since Day 1.

### Abstract classes: a contract every subclass must honour

Sometimes you want to say: *"every payment method must have a `pay()` method — but there's no such thing as a generic payment."* Python's `abc` module (Abstract Base Class) lets you write a class that **can't be created directly** and **forces** its subclasses to fill in the required methods.

```python
from abc import ABC, abstractmethod

class PaymentMethod(ABC):            # ABC = this is an abstract base class
    @abstractmethod
    def pay(self, amount):           # a contract: subclasses MUST define this
        pass

class CashPayment(PaymentMethod):
    def pay(self, amount):
        return f"Paid Rs {amount} in cash"

class CardPayment(PaymentMethod):
    def pay(self, amount):
        return f"Charged Rs {amount} to card"

# We don't care which one it is - abstraction lets us treat them the same
for method in [CashPayment(), CardPayment()]:
    print(method.pay(2000))
```

Two things `abc` buys you. First, you can't create the abstract class itself:

```python
# BREAKS ON PURPOSE - an abstract class is a blueprint with a hole in it
p = PaymentMethod()
```

**TypeError: Can't instantiate abstract class PaymentMethod with abstract method pay.** Second, if a subclass *forgets* to implement `pay()`, Python refuses to build it too — the contract is enforced at creation, not discovered later when something calls `.pay()` and crashes.

## Pillar 3 — Inheritance *(full class on this next time)*

**One class reuses another's code.** Write `class Child(Parent):` and the child gets every attribute and method of the parent for free, then adds its own. No copy-paste, no drift between two near-identical classes.

```python
class Animal:
    def __init__(self, name):
        self.name = name
    def describe(self):
        return f"{self.name} is an animal"

class Dog(Animal):            # Dog inherits everything from Animal
    def bark(self):
        return "Woof"

d = Dog("Tommy")
print(d.describe())          # inherited from Animal, not rewritten
print(d.bark())              # Dog's own method
```

## Pillar 4 — Polymorphism

**"Many shapes": the same method name behaves differently depending on the object.** There are two flavours people ask about in interviews — **overriding** and **overloading**.

### Overriding (runtime polymorphism) — Python does this

A child class **redefines** a method it inherited. Same name, its own version wins. One loop, many behaviours:

```python
class Animal:
    def speak(self):
        return "..."

class Dog(Animal):
    def speak(self):             # OVERRIDES Animal.speak
        return "Woof"

class Cat(Animal):
    def speak(self):             # OVERRIDES Animal.speak
        return "Meow"

for animal in [Dog(), Cat(), Animal()]:
    print(animal.speak())        # same call -> different result per object
```

### Overloading (compile-time polymorphism) — Python does *not*

In Java/C++ you can write two methods with the **same name but different parameters** and the language picks one by the arguments. **Python has no method overloading** — if you define a method twice, the second definition simply *replaces* the first.

```python
# BREAKS ON PURPOSE - this looks like "overloading" but the first add() is gone
class Calculator:
    def add(self, a, b):
        return a + b
    def add(self, a, b, c):      # this REPLACES the two-argument add above
        return a + b + c

calc = Calculator()
print(calc.add(1, 2, 3))         # works
print(calc.add(1, 2))            # crashes: add() now needs three arguments
```

**TypeError: add() missing 1 required positional argument: 'c'.** The two-argument version never survived. The Pythonic way to accept "different numbers of arguments" is **default values** or **`*args`** — one method that flexes, which you already learned in Class 7:

```python
class Calculator:
    def add(self, *numbers):     # *args: any number of arguments
        return sum(numbers)

calc = Calculator()
print(calc.add(1, 2))            # 3
print(calc.add(1, 2, 3))         # 6
print(calc.add(1, 2, 3, 4, 5))   # 15
```
